```python
import json
import os 

with open('config.json', 'r') as f:
    data = json.load(f)

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "hepach2021chimpanzees")
original_data_pathway = os.path.join(pathway, "original_data")

complete_path_1 = os.path.join(original_data_pathway, "cpd_all.txt")
complete_path_2 = os.path.join(original_data_pathway, "cpd_time_locked_all.txt")
complete_path_3 = os.path.join(original_data_pathway, "cpd2_all.txt")
complete_path_4 = os.path.join(original_data_pathway, "cpd2_time_locked_all.txt")

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)
```

```python
import pandas as pd
import numpy as np
import pyreadstat

df1 = pd.read_table(complete_path_1)
df2 = pd.read_table(complete_path_2)
df3 = pd.read_table(complete_path_3)
df4 = pd.read_table(complete_path_4)
experiment_import = [[df1, '1', 'cpd_all'],
                    [df2, '1', 'cpd_time_locked_all'],
                    [df3, '2', 'cpd2_all'],
                    [df4, '2', 'cpd2_time_locked_all']]
for x,y,k in experiment_import:
    x['experiment']=y
    x['experiment_name']=k

```

```python
df2['Food'].replace('B', 'banana', inplace=True)
df2['Food'].replace('G', 'grape', inplace=True)


df3['Food_type'].replace('PE', 'pellet', inplace=True)
df3['Food_type'].replace('GR', 'grape', inplace=True)

df4['Food'].replace('P', 'pellet', inplace=True)
df4['Food'].replace('G', 'grape', inplace=True)
# df4['Food'].unique()

```

```python

df2_event = [['Food','food_in_front_but_not_accessible'], 
             ['Measure1','first_light_on'], 
             ['Board_remove','remove_occluder'], 
             ['Measure2','second_light_on'], 
             ['Food_available','food_within_reach'],
            ['Measure3','light_sequence_initiates_again'], 
           [ 'Remove_scre','screw_removed'], 
           ['Pull_food','food_pulled'], 
           ['Measure4','food_removed_and_light_turned'], 
           ['Put_board','occluder_placed'],
           ['Measure5','post_trial_pupil_measure'], 
           ['Pulls_rope','rope_pulled']]

for x,y in df2_event:
    df2['Event'].replace(x, y, inplace=True)

df2=df2.sort_values(by = ['Subject','Individual_session'])


# output_no_1 = 1
# repeat_1 = 0
# init_var = 0
# temp = []
# for index, row in df2.iterrows():
#     if row['experiment_name'] == 'cpd_time_locked_all':
#         if init_var == 0: 
#             init_var = 1
#             participant_no = row['Subject'] 
#             trial_no = row['Trial'] 
#             temp.append(output_no_1) 
#             continue 
#         if row['Subject'] == participant_no: ## check to see if day still matches
#             if row['Event'] == 'first_light_on':
#                 output_no_1 = output_no_1+1
#             temp.append(output_no_1)
#         else: ##do not need another if statement
#             participant_no = row['Subject']
#             output_no_1 = 1
#             if output_no_1 == 9: ##through 8 sessions
#                 output_no_1 = 1
#             temp.append(output_no_1)
#     else:
#         temp.append('') ## important as there were pretest conditions where session was not applicable
# df2 = df2.assign(trial_timepoints=temp)            

df2_condition = [['NE', 'control-experimenter_pulls_food'],
                  ['NC', 'control-chimpanzee_pulls_food'], 
                  ['CH', 'chimpanzee_helps'], 
                 [ 'EH', 'experimenter_helps'], 
                 ['EG', 'experimenter_helps']]

for x,y in df2_condition:
    df2['Condition'].replace(x, y, inplace=True)

df2.rename(columns={"Robert_present": "stooge_present",
                    "Individual_session":'ses_nr_for_subject'}, inplace=True)
```

```python

df4_event = [['Food','food_in_front_but_not_accessible'], 
             ['Measure1','first_light_on'], 
             ['Board_remove','remove_occluder'], 
             ['Measure2','second_light_on'], 
             ['Food_available','food_within_reach'],
            ['Measure3','light_sequence_initiates_again'], 
           [ 'Remove_scre','screw_removed'], 
           ['Pull_food','food_pulled'], 
           ['Measure4','food_removed_and_light_turned'], 
           ['Put_board','occluder_placed'],
           ['Measure5','post_trial_pupil_measured'], 
           ['Pulls_rope','rope_pulled']]

for x,y in df4_event:
    df4['Event'].replace(x, y, inplace=True)


df4_condition = [['NES', 'non-social_experimenter'],
                ['NE', 'non-social_experimenter'],
                 ['NCS', 'non-social_chimpanzee'],
                  ['NC', 'non-social_chimpanzee'], 
                  ['CS', 'social_chimpanzee'], 
                 [ 'ES', 'social_experimenter']]

for x,y in df4_condition:
    df4['Condition'].replace(x, y, inplace=True)

# df4['Event'].unique()

df4.rename(columns={"Robert_present": "stooge_present",
                    "Individual_session":'ses_nr_for_subject'}, inplace=True)
```

```python
data_frames=[df1, df2, df3, df4]
for index, x in enumerate(data_frames):
    x.columns = map(str.lower, x.columns)
    x=x.applymap(lambda s: s.lower() if type(s) == str else s) 
    x.rename(columns={"subject": "ape"}, inplace=True)
    x['ape'] = x['ape'].str.rstrip()
    x['study_id']="hepach2021chimpanzees"
    data_frames[index]=x
new_df=data_frames[0]
fulldf = pd.concat(data_frames, ignore_index=True, sort=False)
```

```python


fulldf.loc[fulldf.group == 'b', ['participant_2','role','role_2']] = 'jahaga','focal_participant','stooge'
fulldf.loc[fulldf.group == 'a', ['participant_2','role','role_2']] = 'robert','focal_participant','stooge'

```

```python



comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)
fulldf['ape'] = fulldf['ape'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['ape'].replace(x, y, inplace=True)

comp_path_name_errors = os.path.join(pathway_gen, "hepach2021chimpanzees_name_errors.csv")
df_name  = pd.read_csv(comp_path_name_errors)
fulldf['ape'] = fulldf['ape'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['ape'].replace(x, y, inplace=True)

comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)    
fulldf= fulldf.merge(apedf,left_on='ape', right_on='name', how='left')

comp_path_ape_info_2 = os.path.join(pathway_gen, "apes_includeindatabase_2.csv")
apedf_2 = pd.read_csv(comp_path_ape_info_2)
fulldf= fulldf.merge(apedf_2,left_on='participant_2', right_on='name_2', how='left')

# fulldf.columns
fulldf.rename(columns={"ape": "participant",
                       'group':'species_subgroup',
                       "ses_nr_for_subject":'session'}, inplace=True)

fulldf['dyad']=fulldf.participant.str.cat(fulldf.participant_2, sep='_')
```

```python

complete_path_rem = os.path.join(out_pathway, "timepoints_session_discrepancies.csv")
df_rem = pd.read_csv(complete_path_rem)
df_rem = df_rem[~df_rem.suggestion.str.contains("keep")]

rem_list = df_rem[['experiment_name','participant','session','trial']].values.tolist()
```

```python
all_drop=pd.DataFrame()
# Identify rows where both columns contain the string
for rem_row in rem_list:
    experiment_name, participant, session, trial =rem_row
    to_drop = fulldf[(fulldf['experiment_name'] == experiment_name) & 
                    (fulldf['participant'] == participant) &
                    (fulldf['session'] == session) &
                    (fulldf['trial'] == trial)
                    ]
    if len(to_drop)>1:
        print(to_drop)
    all_drop = all_drop.append(to_drop)
    

# print(all_drop)
```
