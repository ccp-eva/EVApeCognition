```python
import json 
import os  

with open('config.json', 'r') as f:
    data = json.load(f)

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "scheumann2006sumatran")
original_data_pathway = os.path.join(pathway, "original_data")

complete_path_1 = os.path.join(original_data_pathway, "Scheumann_2006_exp 1 factors_IJP_SURE_Experiment 1.csv")
complete_path_2 = os.path.join(original_data_pathway, "Scheumann_2006_exp2 factors banana locations_IJP_SURE_Experiment 2.csv")
complete_path_3 = os.path.join(original_data_pathway, "Scheumann_2006_exp2 factors grape locations_IJP_SURE_Experiment 2.csv")

complete_path_4 = os.path.join(original_data_pathway, "Scheumann_2006_exp 1 individual_IJP_SURE_Experiment 1.csv")
complete_path_5 = os.path.join(original_data_pathway, "Scheumann_2006_exp 1 revisits_IJP_SURE_Experiment 1.csv")
complete_path_6 = os.path.join(original_data_pathway, "Scheumann_2006_exp2 individuals_IJP_SURE_Experiment 2.csv")
complete_path_7 = os.path.join(original_data_pathway, "Scheumann_2006_exp2 revisits_IJP_SURE_Experiment 2.csv")

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)
```

```python
exp1 = fulldf[fulldf['experiment'] == '1']
exp1_f = exp1[exp1['experiment_name'] == 'factors']
exp1 = exp1[exp1['experiment_name'] != 'factors']

exp2 = fulldf[fulldf['experiment'] == '2'] 
exp2_f = exp2[exp2['experiment_name'] == 'factors']
exp2 = exp2[exp2['experiment_name'] != 'factors']
```

```python
exp1_ind = exp1[exp1['condition'] == 'kiwi'] ##individual
exp1_ind = exp1_ind.dropna(axis=1, how='all')
exp1_rev = exp1[exp1['condition_r'] == 'kiwi'] ##revisits
exp1_rev = exp1_rev.dropna(axis=1, how='all')
exp1_con = exp1[exp1['condition'] == 'control'] ##control
exp1_con = exp1_con.dropna(axis=1, how='all')

i_list = exp1_ind[['study_id', 'experiment',  'session_day', 'participant','age_in_years',
       'sex', 'species', 'condition', 'number_days_approached',
       'relative_percentage_approached']].values.tolist()
r_list = exp1_rev[[ 'number_revisited',
       'relative_percentage_revisited']].values.tolist()
combined_lol = [lol_1+lol_2 for lol_1,lol_2 in zip(i_list,r_list)]

exp1 = pd.DataFrame(combined_lol, columns=['study_id', 'experiment',  'session_day', 'participant','age_in_years',
       'sex', 'species', 'condition', 'number_days_approached',
       'relative_percentage_approached', 'number_revisited',
       'relative_percentage_revisited'])

full_exp1 = pd.concat([exp1, exp1_con], ignore_index=True, sort=False)
full_exp1 = full_exp1.drop('experiment_name', axis=1)

full_exp1['year']='2002'
full_exp1 = full_exp1[['study_id', 'experiment',  'year', 'participant','age_in_years',
       'sex', 'species', 'session_day','condition', 'number_days_approached',
       'relative_percentage_approached', 'number_revisited',
       'relative_percentage_revisited']]
```

```python
exp2_or = exp2[exp2['condition'] =='orange'] ##individual
exp2_or = exp2_or.dropna(axis=1, how='all')
exp2_ind = exp2[exp2['experiment_name'] =='individual'] ##individual_all
exp2_ind_ba = exp2_ind[exp2_ind['condition'] =='banana'] ##banana
exp2_ind_ba = exp2_ind_ba.dropna(axis=1, how='all')

exp2_ind_gr = exp2_ind[exp2_ind['condition'] =='grape'] ##grape
exp2_ind_gr = exp2_ind_gr.dropna(axis=1, how='all')

exp2_rev = exp2[exp2['experiment_name'] == 'revisits'] ##revisits_all
exp2_rev_ba = exp2_rev[exp2_rev['condition'] =='banana'] ##banana
exp2_rev_ba = exp2_rev_ba.dropna(axis=1, how='all')

exp2_rev_gr = exp2_rev[exp2_rev['condition'] =='grape'] ##grape
exp2_rev_gr = exp2_rev_gr.dropna(axis=1, how='all')

b_ind_list = exp2_ind_ba[['study_id', 'experiment',  'session_day', 'participant','age_in_years',
       'sex', 'species', 'condition', 'number_banana_locations',
       'number_days_approached_banana', 
       'number_days_approached_grape', 'relative_percentage_approached_banana',
       'relative_percentage_approached_grape']].values.tolist()
b_rev_list = exp2_rev_ba[['number_revisited', 'percent_revisited']].values.tolist()

combined_lol_2b = [lol_1+lol_2 for lol_1,lol_2 in zip(b_ind_list,b_rev_list)] ##list comprehension is your friend
# print(combined_lol) 
exp2_banana = pd.DataFrame(combined_lol_2b, columns=['study_id', 'experiment',  'session_day', 'participant','age_in_years',
       'sex', 'species', 'condition', 'number_banana_locations',
       'number_days_approached_banana', 
       'number_days_approached_grape', 'relative_percentage_approached_banana',
       'relative_percentage_approached_grape',
       'number_revisited', 'percent_revisited'])
```

```python

g_ind_list = exp2_ind_gr[['study_id', 'experiment','session_day', 'participant','age_in_years',
       'sex', 'species', 'condition', 
       'number_days_approached_banana', 'number_grape_locations',
       'number_days_approached_grape', 'relative_percentage_approached_banana',
       'relative_percentage_approached_grape']].values.tolist()
g_rev_list = exp2_rev_gr[['number_revisited', 'percent_revisited']].values.tolist()

combined_lol_2g = [lol_1+lol_2 for lol_1,lol_2 in zip(g_ind_list,g_rev_list)] ##list comprehension is your friend
# print(combined_lol) 
exp2_grape = pd.DataFrame(combined_lol_2g, columns=['study_id', 'experiment','session_day', 'participant','age_in_years',
       'sex', 'species', 'condition', 
       'number_days_approached_banana', 'number_grape_locations',
       'number_days_approached_grape', 'relative_percentage_approached_banana',
       'relative_percentage_approached_grape',
       'number_revisited', 'percent_revisited'])
full_exp2 = pd.concat([exp2_banana,exp2_grape, exp2_or], ignore_index=True, sort=False)
full_exp2 = full_exp2.drop('experiment_name', axis=1)

full_exp2['year']='2002'
full_exp2 = full_exp2[['study_id', 'experiment','year', 'participant','age_in_years',
       'sex', 'species', 'session_day','condition', 'number_banana_locations',
       'number_days_approached_banana', 'number_grape_locations',
       'number_days_approached_grape', 'relative_percentage_approached_banana',
       'relative_percentage_approached_grape',
       'number_revisited', 'percent_revisited']]
```
