```python
import json
import os 

with open('config.json', 'r') as f:
    data = json.load(f)
    print(data["publications"])

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "liszkowski2009prelinguistic")
out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)

original_data_pathway = os.path.join(pathway, "original_data")
complete_path_1 = os.path.join(original_data_pathway, "liszkowski_2009_in_on_chimp_data.sav")
# print(out_pathway)
```

```python
import pandas as pd
import numpy as np
import pyreadstat


df = pd.read_spss(complete_path_1, usecols=None, convert_categoricals=True)
complete_out_pathway_csv = os.path.join(original_data_pathway, 'liszkowski_2009_in_on_chimp_data.csv')
df.to_csv(complete_out_pathway_csv, encoding='utf-8-sig', index=False)
# df.to_csv(complete_path_1, encoding='utf-8-sig', index=False)
# print(complete_path_1)
```

```python
df['study_id']="liszkowski2009prelinguistic"
df.columns = map(str.lower, df.columns)
df=df.applymap(lambda s: s.lower() if type(s) == str else s)
df.rename(columns={"individuals": "ape"}, inplace=True)

# original = df.columns.tolist()
# df_columns = pd.DataFrame(original)
complete_out_pathway_column = os.path.join(out_pathway, 'liszkowski2009prelinguisitic_column_update.csv')
# df_columns.to_csv(complete_out_pathway_column, encoding='utf-8-sig', index=False)
```

```python
t1 = df[['study_id', 'ape', 'age', 'cond_order', 'tp_in_t1','tp_on_t1', 'cp_in_t1','cp_on_t1',
    'up_in_t1','up_on_t1', 'dp_in_t1','dp_on_t1', 'tpphase_in_t1', 'tpphase_on_t1',
    'tp_in_t1_lat', 'tp_on_t1_lat']]
t1 = t1.assign(t='1')

t1r = df[['study_id','ape',  'age', 'cond_order', 'tp_in_t1r', 'tp_on_t1r','cp_in_t1r', 'cp_on_t1r',
    'up_in_t1r', 'up_on_t1r','dp_in_t1r', 'dp_on_t1r','tpphase_in_t1r', 'tpphase_on_t1r',
    'tp_in_t1r_lat', 'tp_on_t1r_lat']]
t1r = t1r.assign(t='1r')

t2 = df[['study_id','ape',  'age', 'cond_order', 'tp_in_t2', 'tp_on_t2', 'cp_in_t2', 'cp_on_t2',
    'up_in_t2', 'up_on_t2', 'dp_in_t2', 'dp_on_t2', 'tpphase_in_t2', 'tpphase_on_t2',
    'tp_in_t2_lat', 'tp_on_t2_lat']]
t2 = t2.assign(t='2')

t2r = df[['study_id','ape',  'age', 'cond_order', 'tp_in_t2r', 'tp_on_t2r', 'cp_in_t2r', 'cp_on_t2r',
    'up_in_t2r', 'up_on_t2r', 'dp_in_t2r', 'dp_on_t2r', 'tpphase_in_t2r', 'tpphase_on_t2r',
    'tp_in_t2r_lat', 'tp_on_t2r_lat']]
t2r = t2r.assign(t='2r')

t3 = df[['study_id','ape',  'age', 'cond_order', 'tp_in_t3', 'tp_on_t3','cp_in_t3', 'cp_on_t3',
    'up_in_t3', 'up_on_t3','dp_in_t3', 'dp_on_t3', 'tpphase_in_t3', 'tpphase_on_t3',
    'tp_in_t3_lat', 'tp_on_t3_lat']]
t3 = t3.assign(t='3')

t3r = df[['study_id','ape', 'age', 'cond_order', 'tp_in_t3r', 'tp_on_t3r','cp_in_t3r', 'cp_on_t3r',
    'up_in_t3r', 'up_on_t3r','dp_in_t3r', 'dp_on_t3r', 'tpphase_in_t3r', 'tpphase_on_t3r',
    'tp_in_t3r_lat', 'tp_on_t3r_lat']]
t3r = t3r.assign(t='3r')

t4 = df[['study_id','ape',  'age', 'cond_order', 'tp_in_t4', 'tp_on_t4', 'cp_in_t4', 'cp_on_t4',
    'up_in_t4', 'up_on_t4', 'dp_in_t4', 'dp_on_t4', 'tpphase_in_t4', 'tpphase_on_t4',
    'tp_in_t4_lat', 'tp_on_t4_lat']]
t4 = t4.assign(t='4')

t4r = df[['study_id','ape', 'age', 'cond_order', 'tp_in_t4r', 'tp_on_t4r','cp_in_t4r', 'cp_on_t4r',
     'up_in_t4r', 'up_on_t4r','dp_in_t4r', 'dp_on_t4r', 'tpphase_in_t4r', 'tpphase_on_t4r',
     'tp_in_t4r_lat', 'tp_on_t4r_lat']]
t4r = t4r.assign(t='4r')
```

```python
column_rename  = pd.read_csv(complete_out_pathway_column)
```

```python
data_frames=[t1, t1r, t2, t2r, t3, t3r, t4, t4r]
for index, x in enumerate(data_frames):
    for b,y in zip(column_rename['original'],column_rename['new']):
        x.rename(columns={b:y}, inplace=True)
    data_frames[index]=x
new_df=data_frames[0]
fulldf = pd.concat(data_frames, ignore_index=True, sort=False)
```

```python

complete_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")
```

```python
df_name  = pd.read_csv(complete_path_name_errors)
fulldf['ape'] = fulldf['ape'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['ape'].replace(x, y, inplace=True)
```

```python
complete_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")

apedf = pd.read_csv(complete_path_ape_info)
fulldf = fulldf.merge(apedf,left_on='ape', right_on='name', how='left')
# fulldf.columns
fulldf.rename(columns={"ape": "participant",
                       "age":"age_in_years"}, inplace=True)

space_list = ['cond_order', 'tp_in', 'tp_on', 'cp_in', 'cp_on','tpphase_in', 'tpphase_on']
for x in space_list:
    fulldf[x].replace(' ', '_', inplace=True, regex=True)
```

```python
fulldf.rename(columns={"tp_in": "target_point_in",
                       'tp_on':'target_point_on',
                        'cp_in':'control_point_in',
                        'cp_on':'control_point_on',
                        'up_in':'unspecific_point_in',
                        'up_on':'unspecific_point_on', 
                        'dp_in':'distractor_point_in', 
                        'dp_on':'distractor_point_on', 
                        'tpphase_in':'target_point_phase_in', 
                        'tpphase_on':'target_point_phase_on',
                        'tp_in_lat':'target_point_in_lat', 
                        'tp_on_lat':'target_point_on_lat',
                        't':'trial'}, inplace=True)
```

```python
complete_out_pathway_1 = os.path.join(out_pathway, 'liszkowski2009prelinguistic_standardized.csv')

liszkowski2009prelinguistic_standardized=fulldf[['study_id','participant', 'age_in_years','sex',  'species','trial',
        'cond_order', 'target_point_in', 'target_point_on', 'control_point_in', 'control_point_on',
       'unspecific_point_in', 'unspecific_point_on', 'distractor_point_in', 'distractor_point_on', 'target_point_phase_in', 'target_point_phase_on',
       'target_point_in_lat', 'target_point_on_lat' ]]
liszkowski2009prelinguistic_standardized.to_csv(complete_out_pathway_1, encoding='utf-8-sig', index=False)


complete_out_pathway_2 = os.path.join(out_pathway, 'liszkowski2009prelinguistic_glossary.csv')
names =liszkowski2009prelinguistic_standardized.columns.tolist()
df = pd.DataFrame(names)
df = df.rename(columns={0: "column_name"})
df["description"] = ""
liszkowski2009prelinguistic_glossary=df[["column_name", "description"]]

liszkowski2009prelinguistic_glossary.to_csv(complete_out_pathway_2, encoding='utf-8-sig', index=False)

```
