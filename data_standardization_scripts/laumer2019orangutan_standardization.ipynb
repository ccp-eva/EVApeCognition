```python
import json 
import os  

with open('config.json', 'r') as f:
    data = json.load(f)

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "laumer2019orangutans")
original_data_pathway = os.path.join(pathway, "original_data")

complete_path_1 = os.path.join(original_data_pathway, "Laumer 2019_food preference tests.csv")
complete_path_2 = os.path.join(original_data_pathway, "Laumer 2019_motivation_test.csv")
complete_path_3 = os.path.join(original_data_pathway, "Laumer 2019_quality allocation test.csv")
complete_path_4 = os.path.join(original_data_pathway, "Laumer2019_tool functionality test.csv")
complete_path_5 = os.path.join(original_data_pathway, "Laumer2019_tool sel qual alloc.csv")
complete_path_6 = os.path.join(original_data_pathway, "Laumer2019_tool selection test.csv")

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)
```

```python
import pandas as pd
import numpy as np
import pyreadstat

df1 = pd.read_csv(complete_path_1)
df2 = pd.read_csv(complete_path_2)
df3 = pd.read_csv(complete_path_3)
df4 = pd.read_csv(complete_path_4)
df5 = pd.read_csv(complete_path_5)
df6 = pd.read_csv(complete_path_6)

experiment_import = [[df1, "food_preference_tests", '6'],
                    [df2, "motivation_test", "3"],
                    [df3, "quality_allocation_test", "2"],
                    [df4, "tool_functionality_test", "4"],
                    [df5, "tol_sel_qual_alloc", "5"],
                    [df6, "tool_selection_test", "1"]]
for x,y,k in experiment_import:
    x["experiment_name"]=y
    x["experiment"]=k
```

```python
df3.rename(columns={"Chosen (g=grape, a=apple, p=pellet, b=ball, s=stick)": "chosen_test"}, inplace=True)
# df3.columns
chosen_right=[] 
for index, row in df3.iterrows():
    if row["Option on the right (experimenter's view; g=grape, a=apple, p=pellet, b=ball, s=stick)"] == 'a':
        if row['chosen_test'] == 'p':
            chosen_right.append(row["Option on the left (experimenter's view; g=grape, a=apple, p=pellet, b=ball, s=stick)"])
        else:
            chosen_right.append(row['chosen_test'])
    else:
        chosen_right.append(row['chosen_test'])
df3 = df3.assign(chosen_right=chosen_right)

chosen=[] 
for index, row in df3.iterrows():
    if row["Option on the left (experimenter's view; g=grape, a=apple, p=pellet, b=ball, s=stick)"] == 'a':
        if row['chosen_right'] == 'p':
            chosen.append(row["Option on the right (experimenter's view; g=grape, a=apple, p=pellet, b=ball, s=stick)"])
        else:
            chosen.append(row['chosen_right'])
    else:
        chosen.append(row['chosen_right'])
df3 = df3.assign(chosen=chosen)


# len(chosen)
```

```python
# df4.columns
df4.rename(columns={"Chosen (g=grape, a=apple, p=pellet, b=ball, s=stick)": "chosen_test"}, inplace=True)
chosen=[] 
for index, row in df4.iterrows():
    if row["chosen_test"] == 'p':
        chosen.append(row["Apparatus (b=ball, s=stick)"])
    else:
        chosen.append(row['chosen_test'])
df4 = df4.assign(chosen=chosen)
```

```python
data_frames=[df1, df2, df3,df4,df5,df6]
for index, x in enumerate(data_frames):
    x.columns = map(str.lower, x.columns)
    x=x.applymap(lambda s: s.lower() if type(s) == str else s) 
    x=x.rename(columns={"subject": "ape",
        "food preference test 1-4":"food_preference_test_1-4",
        "option on the right (experimenter's view; g=grape, a=apple, p=(banana) pellet)":"option_on_the_right_experimenter_view",
        "option on the left (experimenter's view; g=grape, a=apple, p=(banana) pellet)":"option_on_the_left_experimenter_view",
        "chosen food (g=grape, a=apple, p=(banana) pellet)":"chosen_food",
        "apparatus (b=ball, s=stick)":"apparatus",
        "option on the right (experimenter's view; p=pellet, b=ball, s=stick)":"option_on_the_right_experimenter_view",
        "option on the left (experimenter's view; p=pellet, b=ball, s=stick)":"option_on_the_left_experimenter_view",
        "chosen (p=pellet, b=ball, s=stick)":"chosen",
        "quality allocation test":"quality_allocation_test",
        "content (g=grape, a=apple, p=pellet)":"content",
        "option on the right (experimenter's view; g=grape, a=apple, p=pellet, b=ball, s=stick)":"option_on_the_right_experimenter_view",
        "option on the left (experimenter's view; g=grape, a=apple, p=pellet, b=ball, s=stick)":"option_on_the_left_experimenter_view",
        "chosen (g=grape, a=apple, p=pellet, b=ball, s=stick)":"chosen",
        "profitable option (yes=1, no=0)":"profitable_option",
        "tool functionality test":"tool_functionality_test",
        "tool selection quality allocation test ":"tool_selection_quality_allocation_test",
        "apparatus on the right (experimenter's view; b=ball, s=stick)":"apparatus_on_the_right_experimenter_view",
        "content left (g=grape, a=apple, p=pellet)":"content_left",
        "option on the right (experimenter's view; b=ball, s=stick)":"option_on_the_right_experimenter_view",
        "option on the left (experimenter's view; b=ball, s=stick)":"option_on_the_left_experimenter_view",
        "chosen (b=ball, s=stick)":"chosen",
        "tool selection test":"tool_selection_test",
        "chosen option (b=ball, s=stick)":"chosen_option",
        "correct tool (yes=1, no=0)":"correct_tool",
        "content right (g=grape, a=apple, p=pellet)":"content_right",
        "apparatus on the left (experimenter's view; b=ball, s=stick)":"apparatus_on_the_left_experimenter_view"})
    # x=x.rename(columns={"":""})
    x['study_id']="laumer2019orangutans"
    data_frames[index]=x
new_df=data_frames[0]
fulldf = pd.concat(data_frames, ignore_index=True, sort=False)
```

```python
code_list=["profitable_option", "correct_tool"]
for index, x in enumerate(code_list):    
    fulldf[x] = fulldf[x].astype(str)
    temp=[]
    for entry in fulldf[x]:
        if entry == '0' or entry=='0.0':
            entry = "no"
        elif entry =='1' or entry=='1.0':
            entry = "yes"
        temp.append(entry)
    fulldf = fulldf.assign(temp_col=temp)
    fulldf=fulldf.rename(columns={'temp_col': x+'_codes'})
```

```python
code_list=["option_on_the_right_experimenter_view", "option_on_the_left_experimenter_view", "apparatus_on_the_left_experimenter_view", "content_right", "chosen_food", "apparatus", "chosen", "content", "apparatus_on_the_right_experimenter_view", "content_left", "chosen_option"]
for index, x in enumerate(code_list):    
    fulldf[x] = fulldf[x].astype(str)
    temp=[]
    for entry in fulldf[x]:
        if entry == 'b':
            entry = "ball"
        elif entry =='s':
            entry = "stick"
        elif entry =='g':
            entry = "grape"
        elif entry =='a':
            entry = "apple"
        elif entry =='p':
            entry = "pellet_banana"
        elif entry =='z':
            entry = "rusk"
        temp.append(entry)
    fulldf = fulldf.assign(temp_col=temp)
    fulldf=fulldf.rename(columns={'temp_col': x+'_codes'})
```

```python
fulldf[['month','day', 'year']] = fulldf['date'].str.split('/',expand=True)
```

```python
comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)
fulldf['ape'] = fulldf['ape'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['ape'].replace(x, y, inplace=True)

comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)   
fulldf= fulldf.merge(apedf,left_on='ape', right_on='name', how='left')
```

```python
fulldf.columns
fulldf.dropna(subset=['ape'], inplace=True)
fulldf.replace('nan', np.nan, inplace=True)

fulldf.rename(columns={"ape": "participant",
                        "apparatus_codes":"apparatus_presented",
                        "chosen_codes":"choice",
                        "content_codes":"food_inside_apparatus"}, inplace=True)
fulldf.columns = fulldf.columns.str.replace('_codes', ' ')
```

```python
fulldf.loc[fulldf.experiment == '3', ['food_inside_apparatus']] = 'banana_pellet'
fulldf.loc[fulldf.experiment == '4', ['food_inside_apparatus']] = 'banana_pellet'
```

```python
update_banana_pellet = ['option_on_the_right_experimenter_view ',
        'option_on_the_left_experimenter_view ', 'choice','food_inside_apparatus',
        'content_right ','content_left ']

for x in update_banana_pellet:
        fulldf[x].replace('pellet_banana', 'banana_pellet', inplace=True, regex=True)
```

```python
comp_path_birth_dates = os.path.join(pathway_gen, "apes_age_calculations.csv")
ape_dob = pd.read_csv(comp_path_birth_dates) 
fulldf= fulldf.merge(ape_dob,left_on='participant', right_on='name', how='left') 
fulldf['dodc'] = fulldf['year'].astype(str) + '-' + fulldf['month'].astype(str) + '-' + fulldf['day'].astype(str)
fulldf['dodc'] = pd.to_datetime(fulldf['dodc'])
fulldf['dob'] = pd.to_datetime(fulldf['dob'])

fulldf['age_in_years'] = (fulldf['dodc'] - fulldf['dob']).dt.days//365
```

```python
fulldf=fulldf[['study_id', 'experiment','experiment_name','year', 'month', 'day', 
        'participant','age_in_years', 'sex', 'species', 'session', 'trial',
        'food_preference_test_1-4','tool_selection_quality_allocation_test',
        'quality_allocation_test', 'tool_functionality_test',
       'tool_selection_test',
       'option_on_the_right_experimenter_view ',
        'option_on_the_left_experimenter_view ', 
        
        'apparatus_presented',
        'apparatus_on_the_right_experimenter_view ','content_right ','apparatus_on_the_left_experimenter_view ', 'content_left ',
        'chosen_food ','choice','chosen_option ','food_inside_apparatus',
        'correct_tool ', 'profitable_option '
         
         ]]
##removed 'option_on_the_right_experimenter_view','option_on_the_left_experimenter_view','apparatus', 'motivation test','chosen',
##'apparatus_on_the_right_experimenter_view', 'apparatus_on_the_left_experimenter_view','chosen_food','content','chosen_option', 'content_right','content_left','correct_tool',

##removed _codes from : 'option_on_the_right_experimenter_view_codes', 'option_on_the_left_experimenter_view_codes', 'apparatus_on_the_right_experimenter_view_codes', 'apparatus_on_the_left_experimenter_view_codes', 'profitable_option_codes', 'chosen_food_codes', 'chosen_codes', 'content_codes', 'chosen_option_codes', 'correct_tool_codes', 'content_right_codes', 'content_left_codes'
```

```python
for index in range(1,7):
    exp = fulldf[fulldf['experiment'] == str(index)]
    exp = exp.dropna(axis=1, how='all')
    comp_out_path = os.path.join(out_pathway, 'laumer2019orangutans_exp'+str(index)+'_standardized.csv')
    exp.to_csv(comp_out_path, encoding='utf-8-sig', index=False)
    names = exp.columns.tolist()
    exp_g = pd.DataFrame(names)
    exp_g = exp_g.rename(columns={0: "column_name"})
    exp_g["description"] = ""
    exp_g=exp_g[["column_name", "description"]]
    comp_out_path_glossary = os.path.join(out_pathway, 'laumer2019orangutans_exp'+str(index)+'_glossary.csv')
    exp_g.to_csv(comp_out_path_glossary, encoding='utf-8-sig', index=False)
```
