```python
import pandas as pd
import numpy as np
import pyreadstat

complete_path_1 = os.path.join(original_data_pathway, "muhlenbeck2017_full_raw.csv")
df = pd.read_csv(complete_path_1)



df['study_id']="muhlenbeck2017cultural"
df.columns = map(str.lower, df.columns)
df=df.applymap(lambda s: s.lower() if type(s) == str else s)

df.rename(columns={"proband": "participant",
                   'age':'age_in_years'}, inplace=True)
```

```python
participant_list = [['ap_vp1','batak'],
        ['ap_vp2', 'bimbo'],
        ['ap_vp3', 'dokana'],
        ['ap_vp4', 'padana'],
        ['ap_vp5', 'pini'],
        ['ap_vp6', 'raja'],    
        ['ap_vp7', 'suaq' ],    
        ['ap_vp8', 'tanah']]
for x,y in participant_list:
    df['participant'].replace(x, y, inplace=True, regex=True)

comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)    
df= df.merge(apedf,left_on='participant', right_on='name', how='left')
```

```python
studyID_standardized=df[['study_id','participant','age_in_years', 'sex','species', 
                     #     'study', 
       'timestamp', 'fixationduration', 'aoiids', 'aoinames', 'fixmark',
       'fixmarknew', 'stimuliname', 'picnumb', 'fixationpointx',
       'fixationpointy',
       #   'file_name', 
       # 'numfix', 'tfixdur', 'meanfixdur',
       # 'sumfixdurmark', 'sumfixdurnmark', 'meanfixdurmark', 'meanfixdurnmark',
       # 'numfixdurmark', 'numfixdurnmark', 'preftest', 'preftestvalid'
       ]]
comp_out_path_stand = os.path.join(out_pathway, 'muhlenbeck2017cultural_standardized.csv')
studyID_standardized.to_csv(comp_out_path_stand, encoding='utf-8-sig', index=False)


names =studyID_standardized.columns.tolist()
df = pd.DataFrame(names)
df = df.rename(columns={0: "column_name"})
df["description"] = ""
studyID_glossary=df[["column_name", "description"]]

comp_out_path_glossary = os.path.join(out_pathway, 'muhlenbeck2017cultural_glossary.csv')
studyID_glossary.to_csv(comp_out_path_glossary, encoding='utf-8-sig', index=False)
```
