```python
import json 
import os  

with open('config.json', 'r') as f:
    data = json.load(f)

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "kanngiesser2010bonobos")
original_data_pathway = os.path.join(pathway, "original_data")

complete_path_1 = os.path.join(original_data_pathway, "Kanngiesser_Call_2010_ exp 1 floor data tab_apes use feature and spatial cues_AnimalCognition.csv")
complete_path_2 = os.path.join(original_data_pathway, "Kanngiesser_Call_2010_ exp 2 table data tab_apes use feature and spatial cues_AnimalCognition.csv")
complete_path_3 = os.path.join(original_data_pathway, "Kanngiesser_Call_2010_ exp 3 table follow up tab_apes use feature and spatial cues_AnimalCognition.csv")

complete_path_4 = os.path.join(original_data_pathway, "table_1_exp1.csv")
complete_path_5 = os.path.join(original_data_pathway, "table_1_exp2.csv")
complete_path_6 = os.path.join(original_data_pathway, "table_1_exp3.csv")

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)
```

```python
import pandas as pd
import numpy as np
import pyreadstat

df1 = pd.read_csv(complete_path_1)
df2 = pd.read_csv(complete_path_2)
df3 = pd.read_csv(complete_path_3)

df11 = pd.read_csv(complete_path_4)
df22 = pd.read_csv(complete_path_5)
df33 = pd.read_csv(complete_path_6)

experiment_import = [[df1, 'floor_data','1'],
                    [df2, 'table_data', '2'],
                    [df3, 'table_follow_up', '3']]
for x,y,k in experiment_import:
    x["experiment_name"]=y
    x['experiment']=k
df1.rename(columns={"sum_correct": "sum_correct_original", "comments": "comments_original"}, inplace=True)
df2.rename(columns={"group": "exp_order"}, inplace=True)
df3.rename(columns={"group": "exp_group"}, inplace=True)
```

```python
df1['choices'].unique()

df1_recode = [['0','nonbaited_container'],
            ['1','baited_container'],
            ['object twice','revisit_Feature_location'],
            ['object','feature'],
            ['distractor twice','revisit_Control_location'],
            ['distractor','control'],
            ['chaos...','coding_impossible'],
            ['x','na'],
            ['?','na']
            ]

for x,y in df1_recode:
    df1['choices'].replace(x, y, inplace=True)
df1['choices'].unique()
```

```python
remdf=["dorien", "natascha", "swela","dunja"]
df1 = df1[~df1.name.isin(remdf)]
```

```python

# df2.loc[df2.name == 'corrie', ['condition_group']] = 'ctrl'
df2 = df2[~df2.name.str.contains("pia")]
# df2.columns
```

```python
df3 = df3[~df3.name.str.contains("ulla")]
# df3['name'].unique()

```

```python

df11['Name_temp'].unique()
df1= df1.merge(df11,left_on='name', right_on='Name_temp', how='left')
# df1.columns
# df1['name'].unique()
```

```python
df2_recode = [['0','incorrect_nonbaited_location'],
            ['1','correct_baited_location'],
            ['missing','trial_not_conducted'],
            ['no choice','no_choice']]

for x,y in df2_recode:
    df2['choice'].replace(x, y, inplace=True)

complete_path_4 = os.path.join(original_data_pathway,'table_1.csv')

# df2['choice'].unique()
df2= df2.merge(df22,left_on='name', right_on='Name_temp', how='left')
```

```python
df3_recode = [['0','search_nonbaited_location'],
            ['1','search_baited_location'],
            ['object','feature']]

for x,y in df3_recode:
    df3['choice'].replace(x, y, inplace=True)

# df3['name'].unique()
```

```python

df3= df3.merge(df33,left_on='name', right_on='Name_temp', how='left')
```

```python
# df1['comments_original'].unique()

sum_comments = ['1', '6', '7', '4', '5', '8','9', '10', '11','2', '3', '12']

code_list = df1[["sum_correct_original", "comments_original"]]
temp = []
for index, row in code_list.iterrows():
    if not pd.isna(row["sum_correct_original"]):
        temp.append(row["sum_correct_original"])
    elif row["comments_original"] in sum_comments: 
        temp.append(row["comments_original"])
    else:
        temp.append("")
df1 = df1.assign(temp_col=temp)
df1=df1.rename(columns={'temp_col': 'sum_correct'})
# df1['sum_correct'].replace('nan', np.nan, inplace=True, regex=True)
# df1['sum_correct'].unique()
```

```python
code_list=["comments_original"]
for index, x in enumerate(code_list):    
    df1[x] = df1[x].astype(str)
    temp=[]
    for entry in df1[x]:
        if entry in sum_comments:
            entry = ""
        if entry == "nan":
            entry = ""
        else:
            entry = entry
        temp.append(entry)
    # print(temp)
    df1 = df1.assign(temp_col=temp)
    df1['temp_col'] = df1["temp_col"].astype(str)
    df1=df1.rename(columns={'temp_col': 'comments'})

# df1['comments'].replace('nan', '', inplace=True, regex=True)
df1['comments_original'].replace('nan', '', inplace=True, regex=True)
# df1['comments'].unique()
```

```python
code_list=["exp_order"]
for index, x in enumerate(code_list):    
    df2[x] = df2[x].astype(str)
    temp=[]
    for entry in df2[x]:
        if entry == "1.0" or entry == "1" or entry == 1 :
            entry = "table_floor"
        elif entry == "2.0" or entry == "2" or entry == 2 :
            entry = "table_only"
        else:
            entry = ""
        temp.append(entry)
    df2 = df2.assign(temp_col=temp)
    df2=df2.rename(columns={'temp_col': x+'_codes'})
```

```python
data_frames=[df1,df2,df3]
for index, x in enumerate(data_frames):
    x.columns = map(str.lower, x.columns)
    x=x.applymap(lambda s: s.lower() if type(s) == str else s) 
    x=x.rename(columns={"name": "ape",
        "species": "species_original",
        "group": "group_original",
        "sum correct":"sum_correct"})
    x['study_id']="kanngiesser2010bonobos"
    data_frames[index]=x
new_df=data_frames[0]
fulldf = pd.concat(data_frames, ignore_index=True, sort=False)
```

```python
comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)
fulldf['ape'] = fulldf['ape'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['ape'].replace(x, y, inplace=True)

comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)   
fulldf= fulldf.merge(apedf,left_on='ape', right_on='name', how='left')
```

```python
fulldf[['month','day', 'year']] = fulldf['date'].str.split('/',expand=True)
fulldf.rename(columns={"ape": "participant"}, inplace=True)

```

```python
comp_path_birth_dates = os.path.join(pathway_gen, "apes_age_calculations.csv")
ape_dob = pd.read_csv(comp_path_birth_dates) 
fulldf= fulldf.merge(ape_dob,left_on='participant', right_on='name', how='left') 
fulldf['dodc'] = fulldf['year'].astype(str) + '-' + fulldf['month'].astype(str) + '-' + fulldf['day'].astype(str)
fulldf['dodc'] = pd.to_datetime(fulldf['dodc'])
fulldf['dob'] = pd.to_datetime(fulldf['dob'])

fulldf['age_in_years'] = (fulldf['dodc'] - fulldf['dob']).dt.days//365
```

```python
fulldf.rename(columns={"group_table1":"condition_group",
                       ' search':"search"}, inplace=True)
# fulldf.columns
```

```python
fulldf=fulldf[[ 'study_id','experiment', 'experiment_name','year','month', 'day',
        'participant', 'age_original','age_in_years', 'sex', 'species', 'session',  'trial',
          'condition','condition_group','exp_group', 'order','food' , 
        'choices','choice', 
       'search', 'sum_correct',   'exp_order_codes','feature','place' ,'comments']]

##remove comments_original and exp_order and position
```

```python
fulldf.columns = fulldf.columns.str.replace('_codes', '')
```

```python
fulldf['food'] = fulldf['food'].str.rstrip()
fulldf.replace(' ', '_', inplace=True, regex=True)
fulldf.replace('&', '_', inplace=True, regex=True)
```

```python
for index in range(1,4):
    exp = fulldf[fulldf['experiment'] == str(index)]
    exp = exp.dropna(axis=1, how='all')
    comp_out_path = os.path.join(out_pathway, 'kanngiesser2010bonobos_exp'+str(index)+'_standardized.csv')
    exp.to_csv(comp_out_path, encoding='utf-8-sig', index=False)
    names = exp.columns.tolist()
    exp_g = pd.DataFrame(names)
    exp_g = exp_g.rename(columns={0: "column_name"})
    exp_g["description"] = ""
    exp_g=exp_g[["column_name", "description"]]
    comp_out_path_glossary = os.path.join(out_pathway, 'kanngiesser2010bonobos_exp'+str(index)+'_glossary.csv')
    exp_g.to_csv(comp_out_path_glossary, encoding='utf-8-sig', index=False)
```
