```python
import json
import os 
import pandas as pd
import numpy as np

with open('config.json', 'r') as f:
    data = json.load(f)

pathway = os.path.join(data["publications"], "sanchez2021chimpanzees")
original_data_pathway = os.path.join(pathway, "original_data")
pathway_gen = os.path.join(data["python_files"])

complete_path_1 = os.path.join(original_data_pathway, "Copy of Sanchez Amaro 2021.csv")

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)

df = pd.read_csv(complete_path_1)

df.columns = map(str.lower, df.columns)
df=df.applymap(lambda s: s.lower() if type(s) == str else s)
df = df.rename(columns={"specie": "species_original"})
df['study_id']="sanchez2021chimpanzees"
```

```python

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)

df = pd.read_csv(complete_path_1)

df.columns = map(str.lower, df.columns)
df=df.applymap(lambda s: s.lower() if type(s) == str else s)
df = df.rename(columns={"specie": "species_original"})
df['study_id']="sanchez2021chimpanzees"
```

```python
df['subject'] = df['subject'].str.rstrip()
df['partner'] = df['partner'].str.rstrip()

pathway_gen = os.path.join(data["python_files"])
comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)

for x,y in zip(df_name['wrong'],df_name['right']):
    df['subject'].replace(x, y, inplace=True)
    df['partner'].replace(x, y, inplace=True)


df['dyad']=df.subject.str.cat(df.partner, sep='_')
```

```python
df['role_2']="partner"
df['role']="subject"
df = df.rename(columns={"subject": "ape", 
    "partner": "ape_2"})
```

```python
comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)  
df= df.merge(apedf,left_on='ape', right_on='name', how='left')

comp_path_ape_info_2 = os.path.join(pathway_gen, "apes_includeindatabase_2.csv")
apedf_2 = pd.read_csv(comp_path_ape_info_2)
df= df.merge(apedf_2,left_on='ape_2', right_on='name_2', how='left')
```

```python
df = df.rename(columns={"left s": "food_on_left_for_focal_participant",
    "left p": "food_on_left_for_partner",
    "right s": "food_on_right_for_focal_participant",
    "right p": "food_on_right_for_partner",
    "choice subject": "choice_focal_participant",
    "choice partner": "choice_partner",
    "subject eat": "focal_participant_eat",
    "partner eat": "partner_eat",
    "subject max": "focal_participant_max",
    "partner max": "partner_max",
    " choice alt": "choice_alternative",
    "bias check left1-right0": "focal_participant_choice_side", 
    "experience_sub":"experience_focal_participant",
    "experience_par":"experience_partner",
    "alternative":"alternative_for_partner"})

# df.columns
df.rename(columns={"ape": "participant", "ape_2":"participant_2"}, inplace=True)
```

```python
replace_list = [["fav_for_resp", "favorite_for_responder"],
                ["fav_for_prop", "favorite_for_proposer"]]
for x,y in replace_list: 
    df['condition'].replace(x, y, inplace=True, regex=True)

df['role'].replace('subject', 'focal_participant', inplace=True, regex=True)
```

```python
sanchez2021chimpanzees_standardized = df[['study_id','year ', 'month',  'day','participant', 'sex','role',
        'participant_2', 'sex_2', 'role_2',
       'species','dyad', 'session', 'trial', 'condition',
        'experience_focal_participant', 'experience_partner',   'leverage',
       'food_on_left_for_focal_participant', 'food_on_left_for_partner', 'food_on_right_for_focal_participant', 'food_on_right_for_partner',
       'alternative_for_partner', 'choice_focal_participant', 'choice_partner', 'focal_participant_eat',
       'partner_eat', 'focal_participant_max', 'partner_max', 'choice_alternative', 
       'focal_participant_choice_side']]

```

```python
comp_out_path_stand = os.path.join(out_pathway, 'sanchez2021chimpanzees_standardized.csv')

sanchez2021chimpanzees_standardized.to_csv(comp_out_path_stand, encoding='utf-8-sig', index=False)
```
