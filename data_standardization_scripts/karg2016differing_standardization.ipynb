```python
import json 
import os  

with open('config.json', 'r') as f:
    data = json.load(f)

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "karg2016differing")
original_data_pathway = os.path.join(pathway, "original_data")

complete_path_1 = os.path.join(original_data_pathway, "Karg_2016_Stick Chess_Daten_Karg_2012.csv")

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv(complete_path_1)

df['study_id']="karg2016differing"
df.columns = map(str.lower, df.columns)
df=df.applymap(lambda s: s.lower() if type(s) == str else s)
```

```python
df = df.rename(columns={"name": "ape"})

comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)

for x,y in zip(df_name['wrong'],df_name['right']):
    df['ape'].replace(x, y, inplace=True)
    df['competitor'].replace(x, y, inplace=True)



df['dyad']=df.ape.str.cat(df.competitor, sep='_')
# df.columns
```

```python
df['role_2']="competitor"
df['role']="focal_participant"
df = df.rename(columns={"competitor": "ape_2"})
```

```python
comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)   
df= df.merge(apedf,left_on='ape', right_on='name', how='left') 

comp_path_ape_info_2 = os.path.join(pathway_gen, "apes_includeindatabase_2.csv")
apedf_2 = pd.read_csv(comp_path_ape_info_2)
df= df.merge(apedf_2,left_on='ape_2', right_on='name_2', how='left')
```

```python
df = df.rename(columns={"group": "group_original",
    "sex_y": "sex",
    "% correct nonsocial pretest": "percent_correct_nonsocial_pretest",
    "% correct nonsocial training": "percent_correct_nonsocial_training",
    "% smaller-looking stick nonsocial test": "percent_smaller_looking_stick_nonsocial_test",
    "% correct social pretest": "percent_correct_social_pretest",
    "% correct social training": "percent_correct_social_training",
    "% smaller-looking stick social test": "percent_smaller_looking_stick_social_test",
    "% smaller-looking stick first condition": "percent_smaller_looking_stick_first_condition",
    "% smaller-looking stick first session": "percent_smaller_looking_stick_first_session",
    "% smaller-looking stick nonsocial session 1": "percent_smaller_looking_stick_nonsocial_session_1",
    "% smaller-looking stick social session 1": "percent_smaller_looking_stick_social_session_1"})
```

```python
# df.columns
df.rename(columns={"ape": "participant", "ape_2":"participant_2",
                   "age":"age_in_years",
                   "rearing history":'rearing_history',
                   "first condition":'first_condition'}, inplace=True)
```

```python
karg2016differing_standardized=df[['study_id',  'participant','age_in_years', 'sex',  'role', 'participant_2', 'sex_2', 'role_2',
        'species','dyad',    'rearing_history',
       'first_condition', 'percent_correct_nonsocial_pretest',
       'percent_correct_nonsocial_training',
       'percent_smaller_looking_stick_nonsocial_test',
       'percent_correct_social_pretest', 'percent_correct_social_training',
       'percent_smaller_looking_stick_social_test',
       'percent_smaller_looking_stick_first_condition',
       'percent_smaller_looking_stick_first_session',
       'percent_smaller_looking_stick_nonsocial_session_1',
       'percent_smaller_looking_stick_social_session_1']]
```

```python
comp_out_path_stand = os.path.join(out_pathway, 'karg2016differing_exp1_standardized.csv')
karg2016differing_standardized.to_csv(comp_out_path_stand, encoding='utf-8-sig', index=False)
```

```python
names =karg2016differing_standardized.columns.tolist()
df = pd.DataFrame(names)
df = df.rename(columns={0: "column_name"})
df["description"] = ""
karg2016differing_glossary=df[["column_name", "description"]]

comp_out_path_glossary = os.path.join(out_pathway, 'karg2016differing_exp1_glossary.csv')
karg2016differing_glossary.to_csv(comp_out_path_glossary, encoding='utf-8-sig', index=False)
```
