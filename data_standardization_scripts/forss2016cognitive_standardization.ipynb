```python
import json 
import os  

with open('config.json', 'r') as f:
    data = json.load(f)

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "forss2016cognitive")
original_data_pathway = os.path.join(pathway, "original_data")

complete_path_1 = os.path.join(original_data_pathway, "raw_data_Data_GLMM_Forss_etal_2016_paper.csv")
complete_path_2 = os.path.join(original_data_pathway, "detour_task_Data_GLMM_exploration_Forss_etal_2016.csv")
complete_path_3 = os.path.join(original_data_pathway, "exploration_Data_GLMM_exploration_Forss_etal_2016.csv")
complete_path_4 = os.path.join(original_data_pathway, "novelty_food_Data_GLMM_exploration_Forss_etal_2016.csv")
complete_path_5 = os.path.join(original_data_pathway, "novel_object_Data_GLMM_exploration_Forss_etal_2016.csv")
complete_path_6 = os.path.join(original_data_pathway, "reversal_learning_Data_GLMM_exploration_Forss_etal_2016.csv")
complete_path_7 = os.path.join(original_data_pathway, "tube_trap_Data_GLMM_exploration_Forss_etal_2016.csv")

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)
```

```python
import pandas as pd
import numpy as np
import pyreadstat

df1 = pd.read_csv(complete_path_1)
df1 = df1.rename(columns={"Task/ Rank":"task_rank"})
df2 = pd.read_csv(complete_path_2)
df3 = pd.read_csv(complete_path_3)
df4 = pd.read_csv(complete_path_4)
df5 = pd.read_csv(complete_path_5)
df6 = pd.read_csv(complete_path_6)
df7 = pd.read_csv(complete_path_7)

experiment_import = [
                    # [df1, 'raw_data',''],
                    [df2, 'detour_task', '1'],
                    [df3, 'honey_task','2'],
                    [df4, 'novel_food', ''],
                    [df5, 'novel_object', ''],
                    [df6, 'reversal_learning_task', '3'],
                    [df7, 'tube_trap_task','7']]
for x,y,k in experiment_import:
    x['task_name']=y
    x['task_rank']=k

task_rename_list = [[1, 'detour_task'],
                    [2,'honey_task_stick_solution'],
                    [3, 'reversal_learning_task'],
                    [4,'honey_task_re-use'],
                    [5,'honey_task_rope_tool'],
                    [6, 'honey_task_rope_solution'],
                    [7, 'tube_trap_task']]
for x, y in task_rename_list:
    df1.loc[df1.task_rank == x, ['task_name']] = y
```

```python
df1_list = df1[['task_name','Individual', 'Sex','Age', 'Species', 'Zoo', 
        '#Zoos Lived in', 'task_rank','Performance', 'Group size']].values.tolist()

combine_list = [['detour_task', df2], ##just for clarity
                ['reversal_learning_task', df6], 
                ['tube_trap_task' ,df7]]

df2_list = df2[['task_name','Individual', 'Sex', 'Age', 'Island', 'TimeApparatus',
       'Latency Solution', 'Bold rate', 'Gentle rate']].values.tolist()
df6_list = df6[['task_name','Individual', 'Sex', 'Age','Island',  'Proportion wrong lids ']].values.tolist()
df7_list = df7[['task_name','Individual',  'Sex', 'Age','Island', 'Rate correct tubes']].values.tolist()
```

```python
for line in df1_list: 
    if line[0] == 'detour_task': 
        # if line[8] == 'Yes':
            for row in df2_list: 
                if line[1] == row[1]:
                    line[0:] += row[0:]
    else:
        pass

df1_df2 = pd.DataFrame(df1_list, columns=['task_name','Individual', 'Sex','Age', 'Species', 'Zoo', 
        '#Zoos Lived in', 'task_rank','Performance', 'Group size', 
        'task_name_2','Individual_2', 'Sex_2', 'Age_2', 'Island', 'TimeApparatus',
       'Latency Solution', 'Bold rate', 'Gentle rate'])
```

```python
df1_df2_list = df1_df2[['task_name','Individual', 'Sex','Age', 'Species', 'Zoo',
        '#Zoos Lived in', 'task_rank','Performance', 'Group size', 'TimeApparatus',
       'Latency Solution', 'Bold rate', 'Gentle rate']].values.tolist()

for line in df1_df2_list: 
    if line[0] == 'reversal_learning_task': 
        # if line[8] == 'Yes':
            for row in df6_list: 
                if line[1] == row[1]:
                    line[0:] += row[0:]
    else:
        pass
df1_df2_df6 = pd.DataFrame(df1_df2_list, columns=['task_name','Individual', 'Sex','Age', 'Species', 'Zoo',
        '#Zoos Lived in', 'task_rank','Performance', 'Group size', 
        'TimeApparatus','Latency Solution', 'Bold rate', 'Gentle rate',
         'task_name_2','Individual_2', 'Sex_2', 'Age_2','Island_2',  'Proportion wrong lids '])
```

```python
df1_df2_df6_list = df1_df2_df6[['task_name','Individual', 'Sex','Age', 'Species', 'Zoo',
        '#Zoos Lived in', 'task_rank','Performance', 'Group size', 
        'TimeApparatus','Latency Solution', 'Bold rate', 'Gentle rate',
        'Proportion wrong lids ']].values.tolist()
for line in df1_df2_df6_list: 
    if line[0] == 'tube_trap_task': 
        # if line[8] == 'Yes':
            for row in df7_list: 
                if line[1] == row[1]:
                    line[0:] += row[0:]
    else:
        pass

df1_df2_df6_df7 = pd.DataFrame(df1_df2_df6_list, columns=['task_name','Individual', 'Sex','Age', 'Island', 'Zoo',
        '#Zoos Lived in', 'task_rank','Performance', 'Group size', 
        'TimeApparatus','Latency Solution', 'Bold rate', 'Gentle rate',
        'Proportion wrong lids ',
        'task_name_3','Individual_3',  'Sex_3', 'Age_3','Island_3', 'Rate correct tubes'])
```

```python
df1_combined = df1_df2_df6_df7[['task_name','Individual', 'Sex','Age', 'Island', 'Zoo',
        '#Zoos Lived in', 'task_rank','Performance', 'Group size', 
        'TimeApparatus','Latency Solution', 'Bold rate', 'Gentle rate',
        'Proportion wrong lids ',
        'Rate correct tubes']]

comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)


data_frames=[df1_combined, df3, df4, df5]
for index, x in enumerate(data_frames):
    x.columns = map(str.lower, x.columns)
    x=x.applymap(lambda s: s.lower() if type(s) == str else s)
    x['study_id']="forss2016cognitive"
    data_frames[index]=x
new_df=data_frames[0]
```

```python
fulldf = pd.concat(data_frames, ignore_index=True, sort=False)
# fulldf.columns
```

```python
fulldf['individual'] = fulldf['individual'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['individual'].replace(x, y, inplace=True)

fulldf = fulldf.rename(columns={"individual": "ape"})
comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info) 
fulldf= fulldf.merge(apedf,left_on='ape', right_on='name', how='left')
```
