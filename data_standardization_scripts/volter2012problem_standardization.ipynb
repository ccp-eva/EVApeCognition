```python
import json
import os 

with open('config.json', 'r') as f:
    data = json.load(f)

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "volter2012problem")
original_data_pathway = os.path.join(pathway, "original_data")

complete_path_1 = os.path.join(original_data_pathway, "Voelter_2012_exp1_AnimCogn_PRCK .csv")
complete_path_2 = os.path.join(original_data_pathway, "Voelter_2012_exp2_AnimCogn_PRCK .csv")
# complete_path_3 = os.path.join(original_data_pathway, "Voelter_2012_exp3_AnimCogn_PRCK .csv")
complete_path_3 = os.path.join(original_data_pathway, "volter2012problem_exp3_standardized_cv.csv")

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)
```

```python
import pandas as pd
import numpy as np
import pyreadstat

df1 = pd.read_csv(complete_path_1)
df1 = df1.assign(experiment='1')
df1.columns
df1_a = df1[['Subject', 'Age',
       'Condition_numeric', 'Condition', 'T1_operation_time',
       'T1_efficiency_first_action', 'Calibration', 'Comments',
       'Sum_T1_T2_operation_time', 'Filter_excluded_subjects',
       'reliability_josep', 'reliability_christoph', 'experiment']]
df1_a = df1_a.assign(T='t1')
df1_b = df1[['Subject', 'Age',
       'Condition_numeric', 'Condition',  'T2_operation_time',
       'T2_efficiency_first_action', 'Calibration', 'Comments',
       'Sum_T1_T2_operation_time', 'Filter_excluded_subjects',
       'reliability_josep', 'reliability_christoph', 'experiment']]
df1_b = df1_b.assign(T='t2')

data_frames=[df1_a, df1_b]
for index, x in enumerate(data_frames):
    x.rename(columns={"T1_operation_time": "operation_time",
        "T2_operation_time":"operation_time",
        "T1_efficiency_first_action":"efficiency_first_action",
        "T2_efficiency_first_action":"efficiency_first_action"}, inplace=True)
    data_frames[index]=x
new_df=data_frames[0]
fulldf1 = pd.concat(data_frames, ignore_index=True, sort=False)

fulldf1.rename(columns={"Filter_excluded_subjects": "success"}, inplace=True)
# fulldf1.columns
```

```python
df2 = pd.read_csv(complete_path_2)
df2 = df2.assign(experiment='2')

df2_a = df2[['Subject',  'Age',
       'Condition', 'Condition_numeric', 'Order', 'Down_T1T2_operation_time',
       'Down_T1_operation_time', 'Down_Solution',
       'Down_T1_Efficiency_First_Action', 'filter', 'success', 'tool_num',
       'Tool', 'Down_T1_.l_sticks', 'Down_T1_.h_sticks', 
        'Down_T1_perc_efficient_sticks',
        'reliability_josep',
       'reliability_christoph', 'reliability_jc_first_action',
        'experiment']]
df2_a = df2_a.assign(T='t1')
df2_b = df2[['Subject',  'Age',
       'Condition', 'Condition_numeric', 'Order', 'Down_T1T2_operation_time',
        'Down_T2_operation_time', 'Down_Solution',
        'filter', 'success', 'tool_num',
       'Tool',  'Down_T2_.l_sticks',
       'Down_T2_.h_sticks',
       'Down_T2_perc_efficient_sticks', 'reliability_josep',
       'reliability_christoph', 'reliability_jc_first_action',
       'Down_T2_Efficiency_First_Action', 'experiment']]
df2_b = df2_b.assign(T='t2')

data_frames=[df2_a, df2_b]
for index, x in enumerate(data_frames):
    x.rename(columns={"Down_T1_operation_time": "down_operation_time",
        "Down_T2_operation_time":"down_operation_time",
        "Down_T1_.l_sticks":"Down_l_sticks",
        "Down_T2_.l_sticks":"Down_l_sticks",
        "Down_T1_.h_sticks":"Down_h_sticks",
        "Down_T2_.h_sticks":"Down_h_sticks",
        "Down_T1_perc_efficient_sticks":"Down_perc_efficient_sticks",
        "Down_T2_perc_efficient_sticks":"Down_perc_efficient_sticks",
        "Down_T1_Efficiency_First_Action":"Down_Efficiency_First_Action",
        "Down_T2_Efficiency_First_Action":"Down_Efficiency_First_Action"}, inplace=True)
    data_frames[index]=x
new_df=data_frames[0]
fulldf2 = pd.concat(data_frames, ignore_index=True, sort=False)
```

```python
data_frames=[fulldf1, fulldf2]
for index, x in enumerate(data_frames):
    x.columns = map(str.lower, x.columns)
    x=x.applymap(lambda s: s.lower() if type(s) == str else s) 
    x.rename(columns={"subject": "ape"}, inplace=True)
    x['ape'] = x['ape'].str.rstrip()
    x['study_id']="volter2012problem"
    data_frames[index]=x
new_df=data_frames[0]
fulldf = pd.concat(data_frames, ignore_index=True, sort=False)
```

```python

comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)
fulldf['ape'] = fulldf['ape'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['ape'].replace(x, y, inplace=True)

comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase_chimpanzee_groups.csv")
apedf = pd.read_csv(comp_path_ape_info)    
fulldf= fulldf.merge(apedf,left_on='ape', right_on='name', how='left')
# fulldf.columns
```

```python
fulldf.rename(columns={"t": "trial",
    "p":"phase"}, inplace=True)
fulldf['trial'].replace('t', '', inplace=True, regex=True)
```

```python
fulldf['tool'] = fulldf['tool'].str.rstrip()
fulldf['comments'] = fulldf['comments'].str.strip()
fulldf['condition'] = fulldf['condition'].str.strip()
fulldf['calibration'] = fulldf['calibration'].str.strip()

fulldf['comments'].replace('\(', '', inplace=True, regex=True)
fulldf['comments'].replace('\)', '', inplace=True, regex=True)

fulldf.rename(columns={"ape": "participant", "filter_excluded_subjects":"drop_out",
                       "age":"age_in_years",
                       "subgroup":"species_subgroup"}, inplace=True)

fulldf=fulldf.sort_values(by = ['trial'])
```
