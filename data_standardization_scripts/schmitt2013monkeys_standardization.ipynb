```python
import json
import os 

with open('config.json', 'r') as f:
    data = json.load(f)

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "schmitt2013monkeys")
original_data_pathway = os.path.join(pathway, "original_data")

complete_path_1 = os.path.join(original_data_pathway, "Schmitt_2013_exp1_AnimCogn_MOSK.csv")
# complete_path_2 = os.path.join(original_data_pathway, "Schmitt_2013_exp2_AnimCogn_MOSK.csv")
complete_path_2 = os.path.join(original_data_pathway, "Data_Size_Discrimination_Exp_2_Schmitt_et_al_2013_new.csv")

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)
```

```python
import pandas as pd
import numpy as np
import pyreadstat

df1 = pd.read_csv(complete_path_1)
df2 = pd.read_csv(complete_path_2)
# df2.columns
```

```python
df1.rename(columns={'name':'participant', 
                               'specie': "species_r",
                               'species':'species_l'}, inplace=True)

df1_left =df1[['Subject', 'time', ' minimal size', 'condition']]
df1_right = df1[['condition', 'participant', 'species_r', 'sex',
       'simul', 'simulp', 'sim_100', 'sim_80', 'sim_60', 'sim_40', 'sim_30',
       'sim_20', 'sim_10', 'sim_5', 'sequen', 'sequenp', 'seq_05', 'seq_20',
       'seq_60', '5 s', '20 s', '60 s']]

df1_left_0 = df1_left[df1_left.time == 0].values.tolist()
df1_left_5 = df1_left[df1_left.time == 5].values.tolist()
df1_left_20 = df1_left[df1_left.time == 20].values.tolist()
df1_left_60 = df1_left[df1_left.time == 60].values.tolist()

combined_lol_df1 = [lol_1+lol_2 for lol_1,lol_2 in zip(df1_left_0,df1_left_5)]
combined_lol_df1 = [lol_1+lol_2 for lol_1,lol_2 in zip(combined_lol_df1,df1_left_20)]
combined_lol_df1 = [lol_1+lol_2 for lol_1,lol_2 in zip(combined_lol_df1,df1_left_60)]

df1_left_temp = pd.DataFrame(combined_lol_df1, columns=['Subject', 'time_0', ' minimal size_0', 'condition_0',
                                                    'Subject_5',  'time_5', ' minimal size_5', 'condition_5',
                                                    'Subject_20',  'time_20', ' minimal size_20', 'condition_20',
                                                    'Subject_60', 'time_60', ' minimal size_60', 'condition_60'])
df1_left_temp = df1_left_temp.sort_values(by = ['Subject'])
df1_left_full = df1_left_temp[['Subject', ' minimal size_0', 'condition_0', ' minimal size_5', 'condition_5',
                                                     ' minimal size_20', 'condition_20',
                                                    ' minimal size_60', 'condition_60']].values.tolist()

df1_right_suc = df1_right[df1_right.simul == 1]
df1_right_suc = df1_right_suc.sort_values(by = ['participant']).values.tolist()

combined_df1 = [lol_1+lol_2 for lol_1,lol_2 in zip(df1_right_suc, df1_left_full)]

df1_success = pd.DataFrame(combined_df1, columns=['condition', 'participant', 'species_r','sex',
       'simul', 'simulp', 'sim_100', 'sim_80', 'sim_60', 'sim_40', 'sim_30',
       'sim_20', 'sim_10', 'sim_5', 'sequen', 'sequenp', 'seq_05', 'seq_20',
       'seq_60', '5 s', '20 s', '60 s','Subject', ' minimal size_0', 'condition_0', ' minimal size_5', 'condition_5',
       ' minimal size_20', 'condition_20',' minimal size_60', 'condition_60'])

df1_right_fail = df1_right[df1_right.simul == 0]

data_frames = [df1_success, df1_right_fail]
fulldf1 = pd.concat(data_frames, ignore_index=True, sort=False)

fulldf1 = fulldf1[['participant', 'sex', 'species_r','condition', 
       'simul', 'simulp', 'sim_100', 'sim_80', 'sim_60', 'sim_40', 'sim_30',
       'sim_20', 'sim_10', 'sim_5', 'sequen', 'sequenp', 'seq_05', 'seq_20',
       'seq_60', '5 s', '20 s', '60 s',' minimal size_0', 'condition_0', ' minimal size_5', 'condition_5',
       ' minimal size_20', 'condition_20',' minimal size_60', 'condition_60']]
fulldf1 = fulldf1.assign(experiment='1') 
```

```python
df2.rename(columns={'Stimulus Set':'Stimulus_Set', 
                    'species':'species_l'}, inplace=True)

df2_left=df2[[ 'Set combined', 'Control', 'Chance']].values.tolist()



df2_right=df2[['spp', 'ID', 'Stimulus_Set', 'Correct']]
df2_right_2a = df2_right[df2_right.Stimulus_Set == '2a'].values.tolist()
df2_right_2b = df2_right[df2_right.Stimulus_Set == '2b'].values.tolist()
combined_lol_df2 = [lol_1+lol_2 for lol_1,lol_2 in zip(df2_right_2a,df2_right_2b)]
df2_right = pd.DataFrame(combined_lol_df2, columns=['species_r', 'participant', 'Stimulus_Set_2a', 'Correct_2a',
                                                    'spp_2b', 'ID_2b', 'Stimulus_Set_2b', 'Correct_2b'])
df2_right= df2_right[['species_r', 'participant', 'Correct_2a',  'Correct_2b']].values.tolist()

combined_df2 = [lol_1+lol_2 for lol_1,lol_2 in zip(df2_right,df2_left)]
fulldf2 = pd.DataFrame(combined_df2, columns=['species_r', 'participant', 'Correct_2a',  'Correct_2b',
                                                     'Set combined', 'Control', 'Chance'])
fulldf2 = fulldf2.assign(experiment='2') 
```

```python
data_frames_full = [fulldf1, fulldf2]
fulldf = pd.concat(data_frames_full, ignore_index=True, sort=False)

fulldf['study_id']="schmitt2013monkeys"
fulldf.columns = map(str.lower, fulldf.columns)
fulldf=fulldf.applymap(lambda s: s.lower() if type(s) == str else s)
```

```python
comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)
fulldf['participant'] = fulldf['participant'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['participant'].replace(x, y, inplace=True)

comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)    
fulldf= fulldf.merge(apedf,left_on='participant', right_on='name', how='left')
```

```python
##need to fix species and sex columns - filling in blanks and removing human participants
spe_2=[]  
for index, row in fulldf.iterrows():
    if not pd.isna(row['species']):
        spe_2.append(row['species'])
    else:
        spe_2.append(row['species_r'])
fulldf = fulldf.assign(species=spe_2)

spe_1=[]  
for index, row in fulldf.iterrows():
    if not pd.isna(row['sex_y']):
        spe_1.append(row['sex_y'])
    else:
        spe_1.append(row['sex_x'])
fulldf = fulldf.assign(sex=spe_1)

fulldf['sex'].replace('female', 'f', inplace=True)
fulldf['sex'].replace('male', 'm', inplace=True)

fulldf['species'].unique()


fulldf = fulldf[~fulldf.species.str.contains('human')]
# fulldf['species'].unique()
```

```python
fulldf.rename(columns={'simul':'simul_pass', 
                       'simulp':"sim_final_percent_volume",
                       'sequen':'sequence_pass',
                       'sequenp':'sequence_final_percent_volume',
                       '5 s':'final_percent_volume_at_5_sequence',
                       '20 s':'final_percent_volume_at_20_sequence',
                       '60 s':'final_percent_volume_at_60_sequence'}, inplace=True)


# fulldf.columns
```

```python
fulldf['participant'].replace('punktjen', 'pünktchen', inplace=True, regex=True)
fulldf['participant'].replace('tropfche', 'tröpfchen', inplace=True, regex=True)
```

```python
complete_path_age = os.path.join(original_data_pathway, "subject_list.csv")
subject_list = pd.read_csv(complete_path_age)   
fulldf= fulldf.merge(subject_list,left_on='participant', right_on='name', how='left')
fulldf.rename(columns={"age": "age_in_years", "set combined":'set_combined'}, inplace=True) 
```

```python
fulldf = fulldf[['study_id','experiment','participant', 'age_in_years','sex', 'species', 'condition', 'simul_pass', 'sim_final_percent_volume',
       'sim_100', 'sim_80', 'sim_60', 'sim_40', 'sim_30', 'sim_20', 'sim_10',
       'sim_5', 'sequence_pass', 'sequence_final_percent_volume', 'seq_05', 'seq_20', 'seq_60', 'final_percent_volume_at_5_sequence',
       'final_percent_volume_at_20_sequence', 'final_percent_volume_at_60_sequence',   'correct_2a', 'correct_2b',
     'set_combined', 'control', 'chance' ]]

##removed ' minimal size_0', 'condition_0', ' minimal size_5',
#       'condition_5', ' minimal size_20', 'condition_20', ' minimal size_60',
#       'condition_60' 
for index in range(1,3):
    exp = fulldf[fulldf['experiment'] == str(index)]
    exp = exp.dropna(axis=1, how='all')
    comp_out_path = os.path.join(out_pathway, 'schmitt2013monkeys_exp'+str(index)+'_standardized.csv')
    exp.to_csv(comp_out_path, encoding='utf-8-sig', index=False)
    names = exp.columns.tolist()
    exp_g = pd.DataFrame(names)
    exp_g = exp_g.rename(columns={0: "column_name"})
    exp_g["description"] = ""
    exp_g=exp_g[["column_name", "description"]]
    comp_out_path_glossary = os.path.join(out_pathway, 'schmitt2013monkeys_exp'+str(index)+'_glossary.csv')
    exp_g.to_csv(comp_out_path_glossary, encoding='utf-8-sig', index=False)
```
