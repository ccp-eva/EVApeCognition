```python
import json 
import os  

with open('config.json', 'r') as f:
    data = json.load(f)

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "bueno2019bargaining")
original_data_pathway = os.path.join(pathway, "original_data")

complete_path_1 = os.path.join(original_data_pathway, "Bueno-Guerra_2019_JCP_BAON_data_Excel.csv")

out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)
```

```python

import pandas as pd
import numpy as np
df = pd.read_csv(complete_path_1)

df['study_id']="bueno2019bargaining"
# df.columns
df = df.replace('#NULL!', np.nan)
```

```python
order_games_codes = []
for entry in df["Order_games"]:
    if entry == '1':
        entry = "DG-UG"
    elif entry == '2':
        entry = "UG-DG"
    order_games_codes.append(entry)
df = df.assign(order_games_codes=order_games_codes)

test1=df[["Order_games", "order_games_codes"]]
```

```python
dyad = []
for entry in df["Dyad_order"]:
    if entry == '1':
        entry = "Lome_Lobo"
    elif entry =='2':
        entry = "Lobo_Lome"
    elif entry == '3':
        entry = "Kofi_Kara"
    elif entry == '4':
        entry = "Kara_Kofi"
    elif entry == '5':
        entry = "Lobo_Kofi"
    elif entry == '6':
        entry = "Kofi_Lobo"
    elif entry == '7':
        entry = "Alex_Jahaga"
    elif entry == '8':
        entry = "Jahaga_Alex"
    dyad.append(entry)
df = df.assign(dyad=dyad)
test1=df[["Dyad_order", "dyad"]]
```

```python
df.Subject = df.Subject.astype(str)
ape = []
for entry in df["Subject"]:
    if entry == '1':
        entry = "lome"
    elif entry =='2':
        entry = "lobo"
    elif entry == '5':
        entry = "lobo"
    elif entry == '6':
        entry = "kofi"
    elif entry == '7':
        entry = "Alex"
    elif entry == '8':
        entry = "Jahaga"
    elif entry == '9':
        entry = "kofi"
    elif entry == '10':
        entry = "kara"
    elif entry == '12':
        entry = "kara"
    ape.append(entry)
df = df.assign(ape=ape)
test1=df[["Subject", "ape"]]
```

```python
df.Subject = df.Subject.astype(str)
temp= []
for entry in df["Responder"]:
    if entry == '1':
        entry = "lome"
    elif entry =='2':
        entry = "lobo"
    elif entry == '5':
        entry = "lobo"
    elif entry == '6':
        entry = "kofi"
    elif entry == '7':
        entry = "alex"
    elif entry == '8':
        entry = "jahaga"
    elif entry == '9':
        entry = "kofi"
    elif entry == '10':
        entry = "kara"
    elif entry == '12':
        entry = "kara"
    temp.append(entry)
df = df.assign(temp_col=temp)
df=df.rename(columns={'temp_col': 'responder_name'})
test1=df[["Subject", "responder_name", "Responder"]]
# df.columns
```

```python
df['responder_name_temp']=df['responder_name']
```

```python
df = df.rename(columns={"responder_name_temp": "ape_2"})

role=[]
role_2=[]
for index, row in df.iterrows():
    if not pd.isna(row['ape']):
        role.append("subject")
    else:
        role.append("")
df = df.assign(role=role)
for index, row in df.iterrows(): 
    if not pd.isna(row['ape_2']):
        role_2.append("responder")
    else:
        role_2.append("")
df = df.assign(role_2=role_2)
```

```python
df.Test_cod = df.Test_cod.astype(str)
test_cod_codes = []
for entry in df["Test_cod"]:
    if entry == '1':
        entry = "Pretest"
    elif entry == '2':
        entry = "DG"
    elif entry == '3':
        entry = "UG"
    elif entry == '4':
        entry = "post test"
    test_cod_codes.append(entry)
df = df.assign(test_cod_codes=test_cod_codes)

test1=df[["Test_cod", "test_cod_codes"]]
```

```python
df.Subtest = df.Subtest.astype(str)
subtest_codes= []
for entry in df["Subtest"]:
    if entry == '1':
        entry = "training_door_close"
    elif entry =='2':
        entry = "training_door_open"
    elif entry == '3':
        entry = "dg"
    elif entry == '4':
        entry = "ug"
    elif entry == '5':
        entry = "postraining1_door_close"
    elif entry == '6':
        entry = "postraining_1_door_open"
    elif entry == '7':
        entry = "postraining_2_door_close"
    elif entry == '8':
        entry = "postraining_2_door_open"
    elif entry == '9':
        entry = "postraining_3_door_close"
    elif entry == '10':
        entry = "postraining_3_door_open"
    subtest_codes.append(entry)
df = df.assign(subtest_codes=subtest_codes)

test1=df[["Subtest", "subtest_codes"]]
```

```python
df.Condition = df.Condition.astype(str)
Condition_codes = []
for entry in df["Condition"]:
    if entry == '1':
        entry = "ncl_(no_cost_low_gift)"
    elif entry == '2':
        entry = "nch_(no_cost_high_gift)"
    elif entry == '3':
        entry = "cl_(cost_low_gift)"
    elif entry == '4':
        entry = "ch_(cost_high_gift)"
    Condition_codes.append(entry)
df = df.assign(Condition_codes=Condition_codes)

test1=df[["Condition", "Condition_codes"]]
```

```python
code_list=["Prop_Choice_labels", "Resp_PreviousChoice_labels", "COM_P1_Quantity", "COM_P2_Quantity", "COM_P3_Quantity", "COM_P4_Quantity"]

for index, x in enumerate(code_list):    
    df[x] = df[x].astype(str)
    temp=[]
    for entry in df[x]:
        if entry == '0':
            entry = "6/0"
        elif entry =='1':
            entry = "6/3"
        elif entry == '2':
            entry = "6/9"
        elif entry == '3':
            entry = "5/3"
        elif entry == '4':
            entry = "5/9"
        elif entry == '5':
            entry = "5/0"
        elif entry == '6':
            entry = "0/0"
        elif entry == '7':
            entry = "0/3"
        else:
            entry=""
        temp.append(entry)
    df = df.assign(temp_col=temp)
    df=df.rename(columns={'temp_col': x+'_codes'})
```

```python

test1=df[["Prop_Choice_labels", "Prop_Choice_labels_codes"]]
```

```python
code_list2=["Relationshp", "Cost", "Amount", "Responder_Choice_UG"]
for index, x in enumerate(code_list2): 
    df[x] = df[x].astype(str)
```

```python
temp = []
for entry in df["Relationshp"]:
    if entry == "0":
        entry = "unrelated"
    elif entry == "1":
        entry = "kinship"
    else:
        entry=""
    temp.append(entry)
df = df.assign(temp_col=temp)
df=df.rename(columns={'temp_col': 'relationship_codes'})
```

```python
test1=df[["Relationshp","relationship_codes"]]
df=df.rename(columns={'Relationshp': 'relationship'})
```

```python
# "Cost": "Cost_0=no cost_1=cost"})
temp = []
for entry in df["Cost"]:
    if entry == '0':
        entry = "no_cost"
    elif entry == '1':
        entry = "cost"
    else:
        entry=""
    temp.append(entry)
df = df.assign(temp_col=temp)
df=df.rename(columns={'temp_col': 'cost_codes'})
```

```python
# "Amount": "Amount_1=low_2=high"
temp = []
for entry in df["Amount"]:
    if entry == '1':
        entry = "low"
    elif entry == '2':
        entry = "high"
    else:
        entry=""
    temp.append(entry)
df = df.assign(temp_col=temp)
df=df.rename(columns={'temp_col': 'amount_codes'})
```

```python
# "Responder Choice UG": "Responder Choice UG_0=not pull reject_1=pull accept"
temp = []
for entry in df["Responder_Choice_UG"]:
    if entry == '0':
        entry = "not_pull_reject"
    elif entry == '1':
        entry = "pull_accept"
    else:
        entry=""
    temp.append(entry)
df = df.assign(temp_col=temp)
df=df.rename(columns={'temp_col': 'Responder_Choice_UG_codes'})
```

```python
# "COM_XX": "COM_XX_0=no_1=yes"})
code_list3=["COM_B1", "COM_P1", "COM_Int1", "COM_P2", "COM_P3", "COM_P4"]
for index, x in enumerate(code_list3):    
    df[x] = df[x].astype(str)
    temp=[]
    for entry in df[x]:
        if entry == '0':
            entry = "no"
        elif entry =='1':
            entry = "yes"
        else:
            entry=""
        temp.append(entry)
    df = df.assign(temp_col=temp)
    df=df.rename(columns={'temp_col': x+'_codes'})
```

```python
# "COM_XX_Role": "COM_XX_Role_1=proposer_2=responder"
code_list4=["COM_B1_Role", "COM_Int1_Role","COM_P1_Role", "COM_P2_Role", "COM_P3_Role", "COM_P4_Role"]
for index, x in enumerate(code_list4):    
    df[x] = df[x].astype(str)
    temp=[]
    for entry in df[x]:
        if entry == '1':
            entry = "proposer"
        elif entry =='2':
            entry = "responder"
        else:
            entry=""
        temp.append(entry)
    df = df.assign(temp_col=temp)
    df=df.rename(columns={'temp_col': x+'_codes'})
```

```python
# "COM_XX_when": "COM_XX_when_1=before choice is made_2=after choice is made"
code_list5=["COM_P1_When", "COM_P2_When", "COM_P3_When", "COM_P4_When"]
for index, x in enumerate(code_list5):    
    df[x] = df[x].astype(str)
    temp=[]
    for entry in df[x]:
        if entry == '1':
            entry = "before_choice_is_made"
        elif entry =='2':
            entry = "after_choice_is_made"
        else:
            entry=""
        temp.append(entry)
    df = df.assign(temp_col=temp)
    df=df.rename(columns={'temp_col': x+'_codes'})
```

```python
df.columns = map(str.lower, df.columns)
df=df.applymap(lambda s: s.lower() if type(s) == str else s)
```

```python
comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)

for x,y in zip(df_name['wrong'],df_name['right']):
    df['ape'].replace(x, y, inplace=True)

comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)   
df= df.merge(apedf,left_on='ape', right_on='name', how='left') 

comp_path_ape_info_2 = os.path.join(pathway_gen, "apes_includeindatabase_2.csv")
apedf_2 = pd.read_csv(comp_path_ape_info_2)
df= df.merge(apedf_2,left_on='ape_2', right_on='name_2', how='left')

# df.columns
```

```python
df = df.replace(['nan'], [''])
df.rename(columns={"ape": "participant", "ape_2":"participant_2"}, inplace=True)
```
