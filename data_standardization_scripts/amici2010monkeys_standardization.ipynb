```python
comp_path_birth_dates = os.path.join(pathway_gen, "apes_age_calculations.csv")
ape_dob = pd.read_csv(comp_path_birth_dates) 

df= df.merge(ape_dob,left_on='participant', right_on='name', how='left') 
df['dodc'] = df['year'].astype(str) + '-' + df['month'].astype(str) + '-' + df['day'].astype(str)
df['dodc'] = pd.to_datetime(df['dodc'])
df['dob'] = pd.to_datetime(df['dob'])

df['age_in_years'] = (df['dodc'] - df['dob']).dt.days//365

df['age_in_months'] = df['dodc'].dt.to_period('M') - df['dob'].dt.to_period('M')
```

```python

df['age_in_months'] = df['age_in_months'].apply(lambda x: np.NaN if pd.isnull(x) else x.n)

df.rename(columns={"baited cup": "baited_cup"}, inplace=True)
```

```python
amici2010monkeys_standardized=df[[ 'study_id', 'experiment_name', 'year', 'month', 'day', 
       'participant','age_in_years', 'sex', 'species', 'session', 'trial',
       'condition', 'baited_cup',
       'chosen_cup_if_choice_differs_from_position_of_reward',
       'correct']]
```

```python
comp_out_path_stand = os.path.join(out_pathway, 'amici2010monkeys_standardized.csv')
amici2010monkeys_standardized.to_csv(comp_out_path_stand, encoding='utf-8-sig', index=False)
```
