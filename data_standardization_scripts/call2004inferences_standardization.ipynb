```python
import json 
import os  

with open('config.json', 'r') as f:
    data = json.load(f)

pathway_gen = os.path.abspath(data["python_files"])
pathway_temp = os.path.abspath(data["publications"])
pathway = os.path.join(pathway_temp, "call2004inferences")
original_data_pathway = os.path.join(pathway, "original_data")

complete_path_1 = os.path.join(original_data_pathway, "Call_2004_noise1.sav")
complete_path_2 = os.path.join(original_data_pathway, "Call_2004_noise2.sav")
complete_path_3 = os.path.join(original_data_pathway, "Call_2004_noise3.sav")
complete_path_4 = os.path.join(original_data_pathway, "Call_2004_noise4.sav")
complete_path_5 = os.path.join(original_data_pathway, "Call_2004_noise5.sav")
complete_path_6 = os.path.join(original_data_pathway, "Call_2004_noise6.sav")


out_pathway = os.path.join(pathway, "standardized_data")
if not os.path.exists(out_pathway):
    os.makedirs(out_pathway)
```

```python
import pandas as pd
import numpy as np
import pyreadstat

df1 = pd.read_spss(complete_path_1, usecols=None, convert_categoricals=True)

df1['experiment']='1'

```

```python
df1.rename(columns={"CHO":"choice"}, inplace=True)
choice_codes = []
for entry in df1["choice"]:
    if entry == 1:
        entry = "left"
    elif entry == 2:
        entry = "right"
    choice_codes.append(entry)
df1 = df1.assign(choice_codes=choice_codes)



out = [i+1 for _ in range(30) for i in range(6) for _ in range(12)]
df1 = df1.assign(session=out)

```

```python
df2 = pd.read_spss(complete_path_2, usecols=None, convert_categoricals=True)
df2['experiment']='2'
df2.rename(columns={"TEST":"condition",
    "CHO":"choice"}, inplace=True)


df2.columns = map(str.lower, df2.columns)
df2=df2.applymap(lambda s: s.lower() if type(s) == str else s)

df2_replace = [['aud both', 'auditory_both'],
                ['aud empty', 'auditory_empty'],
                ['aud full', 'auditory_baited'],
                ['vis both', 'visual_both'],
                ['vis empty', 'visual_empty'],
                ['vis full', 'visual_baited'],
                ['vis ful', 'visual_baited']]
for x,y in df2_replace:
    df2['condition'].replace(x, y, inplace=True, regex=True)


```

```python
df3 = pd.read_spss(complete_path_3, usecols=None, convert_categoricals=True)
df3['experiment']='5'

df3.rename(columns={"TEST":"condition"}, inplace=True)
# condition_codes = []
# for entry in df3["condition"]:
#     if entry == 1:
#         entry = "control"
#     elif entry == 2:
#         entry = "tapping"
#     condition_codes.append(entry)
# df3 = df3.assign(condition_codes=condition_codes)


df3['condition'].replace(1, 'control', inplace=True, regex=True)
df3['condition'].replace(2, 'tapping', inplace=True, regex=True)
```

```python

df4 = pd.read_spss(complete_path_4, usecols=None, convert_categoricals=True)
df4['experiment']='4'

df4.rename(columns={"test":"condition"}, inplace=True)
condition_codes = []
for entry in df4["condition"]:
    if entry == 1:
        entry = "shake_baited_stir_empty"
    elif entry == 2:
        entry = "shake_empty_stir_baited"
    elif entry == 3:
        entry = "control"
    condition_codes.append(entry)
df4 = df4.assign(condition_codes=condition_codes)

# df4.columns

condition_update_list = [['K','shaken_empty'],
                         ['K+','shaken_baited'],
                         ['T','stirred_empty'],
                         ['T+', 'stirred_baited'],
                         ['n','noise'],
                         ['f','food'],
                         ['u','vision'],
                         ['',' ']]


for a,b in condition_update_list: 
    df4.loc[df4.left == a, ['food_on_left_codes']] = b
    df4.loc[df4.right == a, ['food_on_right_codes']] = b


```

```python
df5 = pd.read_spss(complete_path_5, usecols=None, convert_categoricals=True)
df5['experiment']='3'

df5.rename(columns={"test":"condition"}, inplace=True)

df5.columns = map(str.lower, df5.columns)
df5=df5.applymap(lambda s: s.lower() if type(s) == str else s)
df5['condition'].replace('experimental', 'experimental_shaken_vs_rotated', inplace=True, regex=True)



for a,b in condition_update_list: 
    df5.loc[df5.left == a, ['food_on_left_codes']] = b
    df5.loc[df5.right == a, ['food_on_right_codes']] = b
```

```python
df6 = pd.read_spss(complete_path_6, usecols=None, convert_categoricals=True)
df6['experiment']='6'
df6.rename(columns={"test":"condition"}, inplace=True)
df6['condition'].replace('Experimental', 'experimental_recorded_sound', inplace=True, regex=True)
```

```python
data_frames=[ df1, df2, df3, df4, df5, df6]

for index, x in enumerate(data_frames):
    x.columns = map(str.lower, x.columns)
    x=x.applymap(lambda s: s.lower() if type(s) == str else s)
    x = x.rename(columns={"name": "ape",
        "suc": "success",
        "test":"condition",
        "cas":"continuous_trial",
        "left":"food_on_left",
        "right":"food_on_right",
        "cho":"choice"})
    # x['data_subset']="data_subset_" + str(index+1) 
    x['study_id']="call2004inferences"
    data_frames[index]=x
new_df=data_frames[0]
```

```python
fulldf = pd.concat(data_frames, ignore_index=True, sort=False)
# fulldf.columns

choice_codes = []
for entry in fulldf["choice"]:
    if entry == 1:
        entry = "right"
    elif entry == 2:
        entry = "left"
    choice_codes.append(entry)
fulldf = fulldf.assign(choice_codes=choice_codes)
```

```python

# condition_update_list = [['k','shaken_empty'],
#                          ['k+','shaken_baited'],
#                          ['t','stirred_empty'],
#                          ['t+', 'stirred_baited'],
#                          ['n','noise'],
#                          ['f','food'],
#                          ['u','vision']]


# for a,b in condition_update_list: 
#     fulldf.loc[fulldf.food_on_left == a, ['food_on_left_codes']] = b
#     fulldf.loc[fulldf.food_on_right == a, ['food_on_right_codes']] = b

condition_update_list_2 = [['n','yes'],
                        ['k+','yes'],
                        ['t+', 'yes'] ,              
                         ['f','yes'],
                         ['u','yes'],
                         ['k','no'],
                         ['t','no'],  
                         ['','no']
                         ]


for a,b in condition_update_list_2: 
    fulldf['food_on_left'].replace(a, b, inplace=True)
    fulldf['food_on_right'].replace(a, b, inplace=True)
```

```python
comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)
fulldf['ape'] = fulldf['ape'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['ape'].replace(x, y, inplace=True)

comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)   
fulldf= fulldf.merge(apedf,left_on='ape', right_on='name', how='left')
```

```python
# fulldf.columns
fulldf.rename(columns={"ape": "participant"}, inplace=True)

complete_path_age = os.path.join(original_data_pathway, "subject_list.csv")
subject_list = pd.read_csv(complete_path_age)   
fulldf= fulldf.merge(subject_list,left_on='participant', right_on='name', how='left')
fulldf.rename(columns={"age": "age_in_years"}, inplace=True) 
```

```python
fulldf=fulldf[['study_id','experiment', 'participant','age_in_years', 'sex', 'species','session','trial','continuous_trial',  
        'condition',  'food_on_left','food_on_left_codes', 'food_on_right', 'food_on_right_codes', 'choice', 'choice_codes', 
       'success' ]]
```

```python
for index in range(1,7):
    exp = fulldf[fulldf['experiment'] == str(index)]
    exp = exp.dropna(axis=1, how='all')
    comp_out_path = os.path.join(out_pathway, 'call2004inferences_exp'+str(index)+'_standardized.csv')
    exp.to_csv(comp_out_path, encoding='utf-8-sig', index=False)
    names = exp.columns.tolist()
    exp_g = pd.DataFrame(names)
    exp_g = exp_g.rename(columns={0: "column_name"})
    exp_g["description"] = ""
    exp_g=exp_g[["column_name", "description"]]
    comp_out_path_glossary = os.path.join(out_pathway, 'call2004inferences_exp'+str(index)+'_glossary.csv')
    exp_g.to_csv(comp_out_path_glossary, encoding='utf-8-sig', index=False)


```
