```python
# df.rename(columns={"ape": "participant", "ape_2":"participant_2"}, inplace=True)
sallie_rename = ['participant','participant_2','dyad']
for x in sallie_rename:
    df[x].replace('salih', 'sallie', inplace=True, regex=True)

complete_path_age = os.path.join(original_data_pathway, "subject_list.csv")
subject_list = pd.read_csv(complete_path_age)   
df= df.merge(subject_list,left_on='participant', right_on='name', how='left')
df.rename(columns={"age": "age_in_years"}, inplace=True)

complete_path_age = os.path.join(original_data_pathway, "subject_list_2.csv")
subject_list = pd.read_csv(complete_path_age)   
df= df.merge(subject_list,left_on='participant_2', right_on='name_2', how='left')
df.rename(columns={"age_2": "age_in_years_2"}, inplace=True)
```

```python
# df.columns
df.rename(columns={"conditionrecoded": "condition"}, inplace=True)
df['condition'].replace('harm', 'take', inplace=True, regex=True)
df['condition'].replace('neutral', 'give', inplace=True, regex=True)
df['condition'].replace('cont1', 'control_1_no_food', inplace=True, regex=True)
df['condition'].replace('cont2', 'control_2_no_victim', inplace=True, regex=True)
df['condition'].unique()
```

```python
liebal2014does_standardized=df[['study_id',   'participant','age_in_years','sex','role', 
                                'participant_2','age_in_years_2','sex_2','role_2', 'species', 'dyad', 'housing','code_sticktransfer',
       'condition', 'exp_contr',  'towexp_comm',
       'towexpstress', 'towhelpcomm', 'towhelpstress']]
comp_out_path_stand = os.path.join(out_pathway, 'liebal2014does_standardized.csv')
liebal2014does_standardized.to_csv(comp_out_path_stand, encoding='utf-8-sig', index=False)


names =liebal2014does_standardized.columns.tolist()
df = pd.DataFrame(names)
df = df.rename(columns={0: "column_name"})
df["description"] = ""
liebal2014does_glossary=df[["column_name", "description"]]

comp_out_path_glossary = os.path.join(out_pathway, 'liebal2014does_glossary.csv')
liebal2014does_glossary.to_csv(comp_out_path_glossary, encoding='utf-8-sig', index=False)
```
