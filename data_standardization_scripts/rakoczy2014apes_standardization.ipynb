```python

column_rename_path = os.path.join(out_pathway,"rakoczy2014apes_column_update.csv")
column_rename = pd.read_csv(column_rename_path)

data_frames_temp2 = [temp1, temp2, temp3, temp4, temp5, temp6, temp7, temp8, temp9, temp10, temp11, temp12]
for index, x in enumerate(data_frames_temp2):
    for b,y in zip(column_rename['original'],column_rename['new']):
        x.rename(columns={b:y}, inplace=True)
    data_frames_temp2[index]=x
new_df=data_frames_temp2[0]
new_df = pd.concat(data_frames_temp2, ignore_index=True, sort=False)
```

```python
# data_frames=[new_df, df5, df6, df7]
data_frames=[new_df]
for index, x in enumerate(data_frames):
    x.columns = map(str.lower, x.columns)
    x=x.applymap(lambda s: s.lower() if type(s) == str else s) 
    x.rename(columns={"name": "ape", 
        "sex":"sex_original", 
        "group":"group_original", 
        "species":"species_original"}, inplace=True)
    x['study_id']="rakoczy2014apes"
    data_frames[index]=x
# new_df=data_frames[0]
fulldf = pd.concat(data_frames, ignore_index=True, sort=False)
```

```python
fulldf[['group', 'temp']] = fulldf['group_original'].str.split('-',expand=True)
fulldf['group'].replace({'bonobos', 'urangs', 'gorillas'}, np.nan, inplace=True)
```

```python
# fulldf['group'].unique()
```

```python
comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)
fulldf['ape'] = fulldf['ape'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['ape'].replace(x, y, inplace=True)

comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)   
fulldf= fulldf.merge(apedf,left_on='ape', right_on='name', how='left')
```

```python
fulldf.replace('99', '', inplace=True, regex=True)
fulldf.replace(99, '', inplace=True, regex=True)
```

```python
fulldf['trial'].astype(str)
spe_2=[] 
for index, row in fulldf.iterrows():
    if row['trial'] in ["1", "2", "3", "4", "5", "6"]:
        spe_2.append("1")
    elif row['trial'] in ["7", "8", "9", "10", "11", "12"]:
        spe_2.append("2")
    else:
        spe_2.append("")
fulldf = fulldf.assign(session=spe_2)
# fulldf.columns
# fulldf['trial'].unique()
fulldf.rename(columns={"ape": "participant", "group":"subgroup", "pretest_gruppe":'pretest_group',
                       "age":"age_original"}, inplace=True)
# fulldf.columns
fulldf.replace(' ', '_', inplace=True, regex=True)

group_translation = [['nicht_teilgenommen','did_not_participate'],
                    ['teilgenommen',"participated"],
                    ['kein_chimp','not_chimp']]
for x,y in group_translation:    
    fulldf['pretest_group'].replace(x, y, inplace=True, regex=True)
fulldf['session'].unique()
fulldf['session'].replace('', np.nan, inplace=True, regex=True)
```

```python
fulldf['age_original'].replace('', '0', inplace=True, regex=True)
fulldf['age_original'].replace(np.nan, '0', inplace=True, regex=True)
fulldf['age_original']=fulldf['age_original'].astype(int)
# fulldf['age_original'].unique()

fulldf['age_in_years'] = (fulldf['age_original'])//12

fulldf['age_original'].replace(0, np.nan, inplace=True, regex=True)
fulldf['age_in_years'].replace(0, np.nan, inplace=True, regex=True)
```

```python
fulldf.rename(columns={"subgroup": "species_subgroup"}, inplace=True)
remdf_2=["maia"]
fulldf = fulldf[~fulldf.participant.isin(remdf_2)]
```

```python
drop_list = [['1','kibara'],
             ['1','zola'],
             ['1','kara'],
             ['1','swela'],
             ['2','swela']]


drops=[] ##entry appended only when there is a participant (i.e., not na)
for index, row in fulldf.iterrows():
    status = np.nan
    for x,y in drop_list:
        if row['experiment'] == x:
            if row['participant'] == y:
                status = "true"
    drops.append(status)
fulldf = fulldf.assign(drop_out=drops)
```

```python
pretest = fulldf[['study_id', 'experiment','experiment_name',
        'participant','age_original','age_in_years', 'sex','species', 'species_subgroup',
          'session', 'trial', 'all_trials', 
         'pretest_group', 'pref', 'pref_side','drop_out']].copy()
pretest = pretest.assign(condition='pretest') 
pretest.dropna(subset=['pref'], inplace=True)
test = fulldf[['study_id', 'experiment','experiment_name',
        'participant','age_original','age_in_years', 'sex','species', 'species_subgroup',
          'session', 'trial', 'all_trials', 
         'pretest_group', 'decision', 
       'side','drop_out']].copy()
test = test.assign(condition='test') 
test.dropna(subset=['side'], inplace=True)


data_frames2 = [test, pretest]
fulldf_new = pd.concat(data_frames2, ignore_index=True, sort=False)
fulldf_new['trial']=fulldf_new['trial'].astype(int)
```

```python
fulldf_new=fulldf_new[['study_id', 'experiment','experiment_name',
        'participant','age_original','age_in_years', 'sex','species', 'species_subgroup',
          'session', 'trial', 'all_trials', 
          'condition','pref', 'pref_side', 'decision', 
       'side',
        #  'decision1','sub',  'crossed', 'uncrossed', 'blabla',
           'drop_out']]
fulldf_new=fulldf_new.sort_values(by = ['participant','condition','trial'])
```

```python
for index in range(1,5):
    exp = fulldf_new[fulldf_new['experiment'] == str(index)]
    exp = exp.dropna(axis=1, how='all')
    comp_out_path = os.path.join(out_pathway, 'rakoczy2014apes_exp'+str(index)+'_standardized.csv')
    exp.to_csv(comp_out_path, encoding='utf-8-sig', index=False)
    names = exp.columns.tolist()
    exp_g = pd.DataFrame(names)
    exp_g = exp_g.rename(columns={0: "column_name"})
    exp_g["description"] = ""
    exp_g=exp_g[["column_name", "description"]]
    comp_out_path_glossary = os.path.join(out_pathway, 'rakoczy2014apes_exp'+str(index)+'_glossary.csv')
    exp_g.to_csv(comp_out_path_glossary, encoding='utf-8-sig', index=False)
```
