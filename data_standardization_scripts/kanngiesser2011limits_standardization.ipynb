```python
df3["experiment_name"]="scarcity"
df3["experiment"]="3"
df3[['food','split']] = df3['presentation'].str.split(' ',expand=True)

df3['split'] = df3['split'].str.rstrip()
code_list=["split"]
for index, x in enumerate(code_list):    
    df3[x] = df3[x].astype(str)
    temp=[]
    temp2=[]
    for entry in df3[x]:
        if entry == '1' or entry == '12':
            temp.append(entry)
            temp2.append('')
        elif entry == 'left' or entry == 'right':
            temp2.append(entry)
            temp.append('')
        else:
            temp.append('')
            temp2.append('')
    df3 = df3.assign(temp_col=temp)
    df3 = df3.assign(temp_col2=temp2)
    df3=df3.rename(columns={'temp_col': 'number_tubes',
        'temp_col2': 'location'})
```

```python
data_frames=[df1,df2, df3]
for index, x in enumerate(data_frames):
    x.columns = map(str.lower, x.columns)
    x=x.applymap(lambda s: s.lower() if type(s) == str else s) 
    x=x.rename(columns={"name": "ape",    
        "species": "species_original",
        "tested fully?":"tested_fully",
        "give empty tubes back?":"give_empty_tubes_back",
        "returned empty tubes?":"returned_empty_tubes",
        "give tube":"give_tube",
        "time before trade":"time_before_trade"})
    x['study_id']="kanngiesser2011limits"
    data_frames[index]=x
new_df=data_frames[0]
fulldf = pd.concat(data_frames, ignore_index=True, sort=False)
```

```python
exp3_result = ['0', '4', '1', '2', '6', '9', '5','8']

fulldf['result'] = fulldf['result'].str.rstrip()
code_list=["result"]
for index, x in enumerate(code_list):    
    fulldf[x] = fulldf[x].astype(str)
    temp=[]
    temp2=[]
    temp3=[]
    for entry in fulldf[x]:
        if entry == 'banana' or entry == 'pellet' or entry == 'grape' or entry == 'plum' or entry == 'a' or entry == 'b' or entry == 'short':
            temp.append(entry)
            temp2.append('')
            temp3.append(np.nan)
        elif entry == 'trade' or entry == 'no trade':
            temp.append('')
            temp2.append(entry)
            temp3.append(np.nan)
        elif entry in exp3_result:
            temp.append('')
            temp2.append('')
            temp3.append(entry)
        else:
            temp.append('')
            # temp.append(entry)
            temp2.append('')
            temp3.append(np.nan)
    fulldf = fulldf.assign(temp_col=temp)
    fulldf = fulldf.assign(temp_col2=temp2)
    fulldf = fulldf.assign(temp_col3=temp3)
    fulldf=fulldf.rename(columns={'temp_col': 'choice',
        'temp_col2': 'trade',
        'temp_col3': 'trade_number'})

# fulldf['result'].unique()
```

```python
# fulldf['ape'].unique()
##confirmation that missing participant is padana
fulldf['ape'].replace(' ', 'padana', inplace=True)
```

```python

comp_path_name_errors = os.path.join(pathway_gen, "common_name_errors.csv")

df_name  = pd.read_csv(comp_path_name_errors)
fulldf['ape'] = fulldf['ape'].str.rstrip()
for x,y in zip(df_name['wrong'],df_name['right']):
    fulldf['ape'].replace(x, y, inplace=True)

comp_path_ape_info = os.path.join(pathway_gen, "apes_includeindatabase.csv")
apedf = pd.read_csv(comp_path_ape_info)   
fulldf= fulldf.merge(apedf,left_on='ape', right_on='name', how='left')
```

```python
fulldf[['month','day', 'year']] = fulldf['date'].str.split('/',expand=True)
```

```python
fulldf.replace(' ', '_', inplace=True, regex=True)
fulldf['condition'].replace('choice_', 'choice', inplace=True, regex=True)
# fulldf.columns

fulldf.rename(columns={"ape": "participant"}, inplace=True)

comp_path_birth_dates = os.path.join(pathway_gen, "apes_age_calculations.csv")
ape_dob = pd.read_csv(comp_path_birth_dates) 
fulldf= fulldf.merge(ape_dob,left_on='participant', right_on='name', how='left') 
fulldf['dodc'] = fulldf['year'].astype(str) + '-' + fulldf['month'].astype(str) + '-' + fulldf['day'].astype(str)
fulldf['dodc'] = pd.to_datetime(fulldf['dodc'])
fulldf['dob'] = pd.to_datetime(fulldf['dob'])

fulldf['age_in_years'] = (fulldf['dodc'] - fulldf['dob']).dt.days//365
```

```python
fulldf['presentation']=fulldf['presentation'].str.rstrip('_')
# fulldf['presentation'].unique()
```

```python
condition_rename = [['choice', 'initial_choice'], 
                    ['wt_1', 'trading_control_1'], 
                    ['wt_2', 'trading_control_2'], 
                    ['end_b', 'endowment_trial_b'], 
                    ['end_p', 'endowment_trial_p'], 
                    ['ident_b', 'identical_trial_b'], 
                    ['ident_p', 'identical_trial_p'],
                    ['wt_3', 'trading_control_3'], 
                    ['wt_4', 'trading_control_4'], 
                    ['choice1', 'choice_trial_1'], 
                    ['choice2', 'choice_trial_2'], 
                    ['choice3', 'choice_trial_3'], 
                    ['choice4', 'choice_trial_4'],
                    ['fam_1', 'familiarization_trial_1'], 
                    ['fam_2', 'familiarization_trial_2'], 
                    ['fam_3', 'familiarization_trial_3'], 
                    ['ident_a', 'identical_trial_a'], 
                    ['end_a', 'endowment_trial_a'], 
                    ['fam_4', 'familiarization_trial_4'], 
                    ['fam_5', 'familiarization_trial_5'],
                    ['fam_6', 'familiarization_trial_6'], 
                    ['wt_fct_1', 'functional_for_nonfunctional_1'], 
                    ['wt_fct_2', 'functional_for_nonfunctional_2'], 
                    ['tool_wt_fct_1', 'functional_for_nonfunctional_1'], 
                    ['tool_wt_fct_2', 'functional_for_nonfunctional_2']]
for x,y in condition_rename:
    fulldf['condition'].replace(x, y, inplace=True)
# fulldf['condition'].unique()

condition_new = []
for index, row in fulldf.iterrows():
    if row['experiment'] == '2':
        if row['condition'] == 'identical_trial_b':
            condition_new.append('identical_trial_tool_b')
        elif row['condition'] == 'endowment_trial_b':
            condition_new.append('endowment_trial_tool_b') 
        elif row['condition'] == 'identical_trial_a':
            condition_new.append('identical_trial_tool_a') 
        elif row['condition'] == 'endowment_trial_a':
            condition_new.append('endowment_trial_tool_a') 
        else:
            condition_new.append(row['condition'])
    else:
        condition_new.append(row['condition'])
fulldf = fulldf.assign(condition_new=condition_new)

# fulldf['condition'].unique()

fulldf=fulldf.rename(columns={'condition': 'condition_old',
        'condition_new': 'condition'})
```

```python
fulldf=fulldf[['study_id', 'experiment','experiment_name', 'year', 'month', 'day',
       'participant','age_in_years', 'sex','species','trial',
       'condition',  'food', 'tool', 'number_tubes', 'location', 
        'choice', 'trade','trade_number', 'tested_fully', 'give_empty_tubes_back',  
        'returned_empty_tubes', 'comments']]

#removed 'presentation', 'result'
```

```python
for index in range(1,4):
    exp = fulldf[fulldf['experiment'] == str(index)]
    exp = exp.dropna(axis=1, how='all')
    comp_out_path = os.path.join(out_pathway, 'kanngiesser2011limits_exp'+str(index)+'_standardized.csv')
    exp.to_csv(comp_out_path, encoding='utf-8-sig', index=False)
    names = exp.columns.tolist()
    exp_g = pd.DataFrame(names)
    exp_g = exp_g.rename(columns={0: "column_name"})
    exp_g["description"] = ""
    exp_g=exp_g[["column_name", "description"]]
    comp_out_path_glossary = os.path.join(out_pathway, 'kanngiesser2011limits_exp'+str(index)+'_glossary.csv')
    exp_g.to_csv(comp_out_path_glossary, encoding='utf-8-sig', index=False)
```
